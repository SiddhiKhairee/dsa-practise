### Top k frequent elements
Given an integer array nums and an integer k, return the k most frequent elements. You may return the answer in any order.

 

Example 1:

Input: nums = [1,1,1,2,2,3], k = 2

Output: [1,2]

Example 2:

Input: nums = [1], k = 1

Output: [1]

Example 3:

Input: nums = [1,2,1,2,1,2,3,1,3,2], k = 2

Output: [1,2]

 

Constraints:

1 <= nums.length <= 105
-104 <= nums[i] <= 104
k is in the range [1, the number of unique elements in the array].
It is guaranteed that the answer is unique.
 

Follow up: Your algorithm's time complexity must be better than O(n log n), where n is the array's size.

```python
from typing import List
```

```python
nums = [1]
k = 1
hsmp = {}
#create a element:freq hsmp
for i in range(len(nums)):
    if nums[i] in hsmp:
        hsmp[nums[i]] += 1
    else:
        hsmp[nums[i]] = 1

# print(hsmp)

#get all frequencies
freq_count = [v for v in hsmp.values()]
freq_count.sort(reverse=True)

#get top k frequencies
top_k = freq_count[:k]
# print(top_k)

listy = []

#for each top frequency, append the key to the list, delete the key from hashmap
for i in range(len(top_k)):
    key = next((k for k, v in hsmp.items() if v == top_k[i]), None)
    listy.append(key)
    if key in hsmp:
        del hsmp[key]

print(listy)     
```

```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        # create a hashmap of element:frequency
        hsmp ={}
        for i in range(len(nums)):
            if nums[i] in hsmp:
                hsmp[nums[i]] += 1
            else:
                hsmp[nums[i]] = 1

        # get all frequencies and sort them in descending order, then get the top k frequencies as a list
        freq_count = [v for v in hsmp.values()]
        freq_count.sort(reverse=True)
        freq_count = freq_count[:k]

        #create an empty list for output
        listy = []

        for i in freq_count:
            key = next((k for k, v in hsmp.items() if v == i), None) # get the key corresponding to most frequency
            listy.append(key) # add key to output
            del hsmp[key] # delete that key

        return listy
             
```

```python
s = Solution()
result = s.topKFrequent([1,2,1,2,1,2,3,1,3,2], 2)
print(result)
```
