### Fibonacci using meomoization

```python
from typing import List
```

```python
def mystery(n):
    memo = {}

    def solve(n):
        if n <= 1:
            return 1

        # STEP 1:
        # Check whether n is already in memo.
        # If it exists, return its stored value.
        if n in memo:
            return memo[n]


        # STEP 2:
        # Calculate solve(n-1) + solve(n-2).
        # Store that answer in memo[n].
        res = solve(n-1) + solve(n-2)
        memo[n] = res


        # STEP 3:
        # Return the calculated answer.
        return res


    return solve(n)
```
