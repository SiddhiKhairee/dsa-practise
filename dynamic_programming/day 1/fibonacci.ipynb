### Recursively

```python
class Solution:
    def fib(self, n: int) -> int:
        if n == 0:
            return 0
        if n == 1:
            return 1
        else:
            return self.fib(n-1) + self.fib(n-2)
```

### Time and Space cmplexity
For the recursive solution
Every non-base-case call makes two more recursive calls.

The running time follows:
\[
T(n)=T(n-1)+T(n-2)+O(1)
\]

A simple exponential upper bound is \(O(2^n)\). A tighter bound is \(\Theta(\varphi^n)\), where \(\varphi\approx1.618\).

Space complexity is O(n) because the deepest active chain of calls is fib(n) → fib(n-1) → ... → fib(0 or 1). The call stack grows linearly, even though the total number of calls grows exponentially.

### With Memoization

```python
class Solution:
    def fib(self, n: int) -> int:
        memo = {}

        def solve(n):
            if n <= 1:
                return n

            # Check whether n is already cached
            if n in memo:
                return memo[n]

            # Calculate the result recursively
            res = solve(n-1) + solve(n-2)
            

            # Store the result in memo
            memo[n] = res

            # Return the answer
            return res

        return solve(n)
```

### Time and Space Complexity
Complexity
- Time: O(n) — Each distinct Fibonacci state is computed once.
- Space: O(n) — O(n) for the memo dictionary and O(n) for the recursive call stack.

### Using Tabulation

```python
class Solution:
    def fib(self, n: int) -> int:

        # Handle small inputs
        if n == 0:
            return 0

        # Create dp array of size n + 1
        dp = [0]*(n+1)

        # Initialize the base cases
        dp[0] = 0
        dp[1] = 1

        # Use a loop to calculate dp[2] through dp[n]
        for i in range(2, n+1):
            dp[i] = dp[i-1] + dp[i-2]

        # Return the final answer
        return dp[n]
```

### Time and Space Complexity

Your bottom-up solution has:
- Time: O(n) — We iterate from 2 to n, performing constant work per iteration.
- Space: O(n) — We store n + 1 elements in our DP array.
