### Diameter of the binaary tree (again)

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:

        # we introduce a nonlocal variable called best
        # this best currently has the value = 0
        best_diameter = 0

        # we then define a dfs function
        def dfs(node):

            nonlocal best_diameter
            # if node does not exist, then we return 0, that means there is no diameter of the tree
            if node is None:
                return 0

            # if node does exist, then we find the height from this node to the leaf, on the left subtree and the right subtree
            # then we add up the heights of the left and the right to find th best possible diameter

            left_height = dfs(node.left)
            right_height = dfs(node.right)

            best_diameter = max(best_diameter, left_height + right_height) # keep on updating the diameter value

            return 1 + max(left_height, right_height)

        dfs(root)
        return best_diameter
```

Time and space complexity
Metric	Complexity	Explanation

Time	O(n)	Each node is visited once

Auxiliary space	O(h)	Recursion stack depends on tree height

Balanced tree space	O(log n)	Logarithmic height

Skewed tree space	O(n)	Linear height
