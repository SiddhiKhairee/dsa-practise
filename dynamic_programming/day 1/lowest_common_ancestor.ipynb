### Lowest Common Ancestor of a Binary Tree

```python
from typing import List
```

```python
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':

        def dfs(node, p, q) -> TreeNode:

            # say the node we have recieved is none. return None
            if node is None:
                return None

            # if node val == p val
            if node.val == p.val:
                return p
            
            # if node val == q val
            if node.val == q.val:
                return q

            # if none of the above cases work, then we call dfs on left and right subtree of the node
            left = dfs(node.left, p, q)
            right = dfs(node.right, p, q)

            # if both left and right return node values, means we found p in one subtree, and q in another subtree and thus we return the node as the LCA
            if left and right:
                return node

            # if left is not none, right is none
            if left is not None and right is None:
                return left
            
            if left is None and right is not None:
                return right

        return dfs(root, p, q)




        
            
```

Time complexity: O(n)

In the worst case, your recursive DFS visits every node in the binary tree.

Auxiliary space: O(h)

The recursion stack depends on the tree's height.
- Balanced tree: O(log n).
- Skewed tree: O(n).
No visited set or memoization is needed.
