### Combination Sum

```python
from typing import List
```

```python
class Solution:
    def combinationSum(self, candidates: list[int], target: int) -> list[list[int]]:

        output = []

        def dfs(index, path, current_sum):

            # when do we prune
            if current_sum > target:
                return

            # when do we add path to the output array
            if current_sum == target:
                output.append(path.copy())
                return

            # otherwise we do a for loop
            for i in range(index, len(candidates)):

                # CHOOSE
                path.append(candidates[i])

                # EXPLORE
                dfs(i, path, current_sum + candidates[i])

                # UNDO
                path.pop()

        dfs(0, [], 0)
        return output
```

### Without the for Loop

```python
class Solution:
    def combinationSum(self, candidates: list[int], target: int) -> list[list[int]]:

        output = []

        def dfs(index, path, current_sum):

            # PRUNE
            if current_sum > target:
                return

            # Base case: add to output
            if current_sum == target:
                output.append(path.copy())
                return

            # Base Case: when is index invalid
            if index == len(candidates):
                return

            # CHOICE 1: Include current candidate
            path.append(candidates[index])
            dfs(index, path, current_sum+candidates[index])
            path.pop()

            # CHOICE 2: Exclude the current candidate and move onto the next index without adding anything to the path
            dfs(index+1, path, current_sum)

        dfs(0, [], 0)
        return output
```
