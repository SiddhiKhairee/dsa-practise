### House Robber - II


You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed. All houses at this place are arranged in a circle. That means the first house is the neighbor of the last one. Meanwhile, adjacent houses have a security system connected, and it will automatically contact the police if two adjacent houses were broken into on the same night.

Given an integer array nums representing the amount of money of each house, return the maximum amount of money you can rob tonight without alerting the police.

 

Example 1:

Input: nums = [2,3,2]
Output: 3
Explanation: You cannot rob house 1 (money = 2) and then rob house 3 (money = 2), because they are adjacent houses.
Example 2:

Input: nums = [1,2,3,1]
Output: 4
Explanation: Rob house 1 (money = 1) and then rob house 3 (money = 3).
Total amount you can rob = 1 + 3 = 4.
Example 3:

Input: nums = [1,2,3]
Output: 3

what we do different in this case is that we use the same function as house robber 1 but we do it on two different subrarrays

subarray 1: will include the first element 0 upto element n-1 and get the value of the max robbery
subarray 2: will include the nth element upto the second element aka element 1, and the value of the max robbery
we return whichever subarray has max value of robbery

### recursive solution

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        n = len(nums)
        if n == 1:
            return nums[0]

        nums_with_first_element = nums[:n-1] # both subarrays have the length of n-1
        nums_with_last_element = nums[1:]

        # at the ith position, do we rob this house or do we skip it
        def solve(i, arr):

            # base case, if i is out of bounds return 0
            if i<0 or i>=n-1:
                return 0

            pick_i = arr[i] + solve(i+2, arr)
            skip_i = solve(i+1, arr)
            return max(pick_i, skip_i)


        return max(solve(0, nums_with_first_element), solve(0, nums_with_last_element))
```

### Memoization

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        n = len(nums)
        if n == 1:
            return nums[0]
        nums_with_first_element = nums[:n-1] # both subarrays have the length of n-1
        nums_with_last_element = nums[1:]

        def solve(i, arr):
            memo = {}
            def recursive_calling(i): 
                # out of bounds i
                if i<0 or i>=n-1:
                    return 0

                if i in memo:
                    return memo[i]

                pick_i = arr[i] + recursive_calling(i+2)
                skip_i = recursive_calling(i+1)
                res = max(pick_i, skip_i)
                memo[i] = res
                return res
            return recursive_calling(i, arr)

        return max(solve(0,nums_with_first_element), solve(0, nums_with_last_element))
```

### Tabulation

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        n = len(nums)
        if n==1:
            return nums[0]
        nums_with_first_element = nums[:n-1]
        nums_with_last_element = nums[1:]

        def solve(arr):
            
            dp = [0]*(len(arr)+2)
            for i in range(len(arr)-1, -1, -1):
                dp[i] = max(arr[i]+dp[i+2], dp[i+1])
            return dp[0]

        return max(solve(nums_with_first_element), solve(nums_with_last_element))
```

### Space Optimized

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        n = len(nums)
        if n==1:
            return nums[0]
        nums_with_first_element = nums[:n-1]
        nums_with_last_element = nums[1:]

        def solve(arr):
            a = 0
            b = 0
            for i in range(len(arr)-1, -1, -1):
                ith_val = max(arr[i]+b, a)
                b = a
                a = ith_val

            return a

        return max(solve(nums_with_first_element), solve(nums_with_last_element))
```
