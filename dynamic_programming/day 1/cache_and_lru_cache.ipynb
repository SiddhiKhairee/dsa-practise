### Cache and Lru Cache in Python

```python
from typing import List
```

```python
# Fibonacci using cache
from functools import cache

@cache
def fib(n):
    if n<=1:
        return n
    return fib(n-1) + fib(n-2)
```

Complexity with @cache
Metric	Complexity

Time	O(n)

Cache space	O(n)

Recursion stack	O(n)

Total auxiliary space	O(n)

Important: @cache does not remove recursion. It eliminates repeated calculations of previously computed states.

```python
# using LRU Cache
from functools import lru_cache
@lru_cache(maxsize=9)
def fib(n):
    if n<=1:
        return n
    return fib(n-1) + fib(n-2)
```

```python
### climbing stairs using Cache
from functools import cache
def climbing_stairs(n):
    @cache
    def solve(n):
        if n<=2:
            return n
        return solve(n-1) + solve(n-2)

    return solve(n)
```
