### Same Tree (Again)

```python
from typing import List
```

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def is_same_tree(p:Node, q:Node) -> bool:

    # we use dfs to find out whether it is the same tree or not
    # we start off by definining a couple base cases where values could be false and immediately returned
    # if p and q, both are none, then we have to return true
    if p is None and q is None:
        return True
    # if one of them is none
    if p is None and q is not None:
        return False
    if q is None and p is not None:
        return False
    # if both are not none
    if p and q:
        # check the values
        if p.val != q.val:
            return False


    # now we recursively compare their left subtrees
    left_same = is_same_tree(p.left, q.left)
    right_same = is_same_tree(p.right, q.right)
    return left_same and right_same

    
```

Time: O(n + m), where n and m are the numbers of nodes in the two trees. In the worst case, we examine every node.



Auxiliary space: O(h), where h is the maximum recursion depth reached. This is O(log n) for balanced trees of comparable size and O(n) in the worst case for a skewed tree.
