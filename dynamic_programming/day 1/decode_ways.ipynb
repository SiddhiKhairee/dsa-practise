### Decode Ways

### Code using recursion

```python
from typing import List
```

```python
class Solution:
    def numDecodings(self, s: str) -> int:

        # create a variable ways, in which we keep on updating the number of ways to decode a string
        

        # we create a dfs function
        # this function will recieve an index value = i
        # barring that index, it will tell us the number to ways to decode th rest of the string
        # so in how many ways would we be able to dcode s[i:]

        def dfs(i) -> int:

            
            # base case number 1, if i reaches the end, it means there is atleast 1 way to decode the string
            if i == len(s):
                return 1

            # base case number 2, if i is pointing at the character '0', then there is no possible decoding for i
            if s[i] == '0':
                return 0

            ways = 0
            # say we gotta take the branch of only next 1 digit
            ways += dfs(i+1)

            # now we need to take the choice of picking the next two digits
            # this can only happen in the following condition
            if (i < len(s)-1 and (s[i] == '1') or (i < len(s)-1 and s[i] == '2' and s[i+1] in '0123456')):
                
                ways += dfs(i+2)

            return ways

        return dfs(0)
        
        
```

### Memoization

```python
from functools import cache
class Solution:
    
    def numDecodings(self, s: str) -> int:

        @cache
        def dfs(i):

            if i == len(s):
                return 1
            if s[i] == '0':
                return 0
            ways = 0
            ways += dfs(i+1)

            if ((i<len(s)-1 and s[i] == '1') or (i<len(s)-1 and s[i] == '2' and s[i+1] in '0123456')):
                ways += dfs(i+2)

            return ways

        return dfs(0)  
```

### Tabulation

```python
class Solution:
    def numDecodings(self, s: str) -> int:
        n = len(s)
        # make a dp
        dp = [0]*(n+1)
        dp[n] = 1 #base case
        

        for i in range(n-1, -1, -1):

            if s[i] != '0':
                dp[i] = dp[i+1]

            if ((i<len(s)-1 and s[i] == '1') or (i<len(s)-1 and s[i] == '2' and s[i+1] in '0123456')):
                dp[i] += dp[i+2]

        return dp[0]

             
```

### Space Optimization

```python
class Solution:
    def numDecodings(self, s: str) -> int:
        n = len(s)

        next1 = 1  # dp[n]
        next2 = 0  # No dp[n+1] state needed

        for i in range(n - 1, -1, -1):
            current = 0

            # Choice 1: Decode one digit
            if s[i] != '0':
                current += next1

            # Choice 2: Decode two digits
            if (
                i < n - 1
                and (
                    s[i] == '1'
                    or (s[i] == '2' and s[i+1] in '0123456')
                )
            ):
                current += next2

            # Shift variables for the next iteration
            next2 = next1
            next1 = current

        return next1
```
