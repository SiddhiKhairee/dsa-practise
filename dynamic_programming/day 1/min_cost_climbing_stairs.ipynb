### Min Cost Climbing Stairs 
You are given an integer array cost where cost[i] is the cost of ith step on a staircase.

Once you pay the cost, you can either climb one or two steps.

You can either start from the step with index 0, or the step with index 1.

Return the minimum cost to reach the top of the staircase, which is the position just past the last step (index cost.length).

```python
from typing import List
```

```python
class Solution:
    def minCostClimbingStairs(self, cost: list[int]) -> int:
        n = len(cost) 
        def cost_for_ith_stair(i):
            if i == 0 or i == 1:
                return 0

            return min(cost_for_ith_stair(i-1)+cost[i-1], cost_for_ith_stair(i-2)+cost[i-2])

        return cost_for_ith_stair(n)  
```

### Memoization

```python
class Solution:
    def minCostClimbingStairs(self, cost: list[int]) -> int:
        # step 1: create an empty dictionary
        memo = {}
        # define an internal function which will calculate the cost to reach the ith stair
        def cost_for_ith_stair(i):

            # define base cases
            if i == 0 or i == 1:
                return 0
            # after that we check if i is present in the memo table
            # if it is, we simply retrieve its value
            # if it isnt, then we use recursion, and then add the value to the table
            if i in memo:
                return memo[i]
            res = min(cost_for_ith_stair(i-1)+cost[i-1] , cost_for_ith_stair(i-2)+cost[i-2])
            memo[i] = res
            return res

        # let n be the length of cost
        n = len(cost)
        # this means that cost currently has the price for n-1 steps
        # for the nth step aka top, we simply call the function
        return cost_for_ith_stair(n)
        
```

### Tabulization

```python
class Solution:
    def minCostClimbingStairs(self, cost: list[int]) -> int:

        n = len(cost)
        # simply initialize the dp
        dp = [0]*(n+1)
        # we know that dp[0] and dp [1] must be 0, so we never update them
        for i in range(2, n+1):
            dp[i] = min(dp[i-1]+cost[i-1] , dp[i-2]+cost[i-2])

        return dp[n]
        
```

### Space optimization
        


```python
class Solution:
    def minCostClimbingStairs(self, cost: list[int]) -> int:
        n = len(cost)

        a = 0
        b = 0

        for i in range(2, n+1):
            cost_to_reach_i = min(a+cost[i-2], b+cost[i-1])
            a = b
            b = cost_to_reach_i

        return b     
```
