### Maximum Depth Of Binary Tree (Again)

```python
from typing import List
```

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def max_depth(node:Node) -> int:

    # base case, if the node does not exist, we return 0
    if node is None:
        return 0

    # then as we traverse the levels at each level we gotta add one
    # we traverse down the left subtree, and the right subtree, return the max tree height and then add 1 for the node
    left_subtree = max_depth(node.left)
    right_subtree = max_depth(node.right)
    return 1 + max(left_subtree, right_subtree)
```

2. Time and space complexity
Let's consider a tree containing n nodes.

Time complexity: O(n)

Every node is visited exactly once, and the work performed at each node is constant.

Auxiliary space: O(h)

The recursive call stack can contain one call for every node along the longest root-to-leaf path.
