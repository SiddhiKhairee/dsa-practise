### Climbing Stairs

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

 

Example 1:

Input: n = 2
Output: 2
Explanation: There are two ways to climb to the top.
1. 1 step + 1 step
2. 2 steps
Example 2:

Input: n = 3
Output: 3
Explanation: There are three ways to climb to the top.
1. 1 step + 1 step + 1 step
2. 1 step + 2 steps
3. 2 steps + 1 step

### Recursive Solution

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        if n == 1:
            return 1
        if n == 2:
            return 2
        return self.climbStairs(n-1) + self.climbStairs(n-2)
```

Your naive recursive implementation has:
- Time: O(2ⁿ) — An exponential upper bound because it repeatedly solves the same subproblems.
- Space: O(n) — The maximum recursive call-stack depth is proportional to n.

### Memoization

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        memo = {}
        def solve(n):
            if n <= 2:
                return n

            if n in memo:
                return memo[n]
            
            res = solve(n-1) + solve(n-2)
            memo[n] = res

            return res
        return solve(n)
            
```

Time: O(n)

There are at most n distinct states. Each state is fully calculated once, with constant work outside the recursive calls.

Space: O(n)
- Memo dictionary: O(n)
- Recursive call stack: O(n)

### Tabulation

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        # Step 1: if input is 1 or 2
        if n<=2:
            return n

        # make an output array
        output = [0]*(n+1)
        # initialize the base cases within the output array
        output[0] = 0
        output[1] = 1
        output[2] = 2

        for i in range(3, n+1):
            output[i] = output[i-1] + output[i-2]

        return output[n]
```

Time: O(n) — Your loop calculates each DP state once.

Space: O(n) — The output array stores n + 1 values.

### Optimized Space Complexity Approch

```python
class Solution:
    def climbStairs(self, n: int) -> int:

        # base cases
        if n<=2:
            return n

        # intitialize 2 variables
        a = 1
        b = 2

        for i in range(3, n+1):
            ith_value = a+b
            a = b
            b = ith_value

        return b
```
