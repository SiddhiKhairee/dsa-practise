### Balanced Binary Tree

```python
from typing import List
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def isBalanced(self, root: TreeNode | None) -> bool:

        # what our function will do is calculate the max depth of the left subtree
        # and calculate the max depth of the right subtree
        # if their max depth differs > 1, return false
        
        # we initially define a get_depth function, which will take a node and return its max depth
        def get_depth(node: TreeNode):
            if node is None:
                return (0, True)
            left_height, balanced_left = get_depth(node.left)
            right_height, balanced_right = get_depth(node.right)

            return (1 + max(left_height, right_height), abs(left_height-right_height)<=1 and balanced_left and balanced_right)

        if root is None: # the root doesn't exist, ofcourse the tree is balanced
            return True

        _ , is_balanced = get_depth(root)

        return is_balanced      
```
