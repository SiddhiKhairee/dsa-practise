### Invert Binary Tree (Again)

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def invertTree(self, root: TreeNode | None) -> TreeNode | None:

        # base condition
        if root is None:
            return None

        # say we have a root and there are left subtrees and right subtrees
        # so for each root, what we would do is return the root with the trees swapped
        left_subtree = self.invertTree(root.left)
        right_subtree = self.invertTree(root.right)

        root.left = right_subtree
        root.right = left_subtree
        return root
```

Time and space complexity
Time complexity: O(n)

Each node is visited once. At each node, you do constant work: two pointer assignments.

Auxiliary space: O(h)

The recursion stack grows according to the height of the tree.
- Balanced tree: O(log n).
- Skewed tree: O(n).
