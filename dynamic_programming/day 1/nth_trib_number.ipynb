### Nth Tribonacci Number


```python
from typing import List
```

```python
def nth_tribonacci_number(n: int) -> int:
    # WRITE YOUR BRILLIANT CODE HERE
    if n<=1:
        return n

    a = 0
    b = 1
    c = 1
    for i in range(3, n+1):
        ith_value = a + b + c
        a = b
        b = c
        c = ith_value
    return c
```

```python
path = []

def dfs(i):
    if i == 2:
        return

    path.append(i)
    dfs(i + 1)
    path.pop()

dfs(0)
print(path)
```
