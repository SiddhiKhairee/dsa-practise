### House Robber

You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed, the only constraint stopping you from robbing each of them is that adjacent houses have security systems connected and it will automatically contact the police if two adjacent houses were broken into on the same night.

Given an integer array nums representing the amount of money of each house, return the maximum amount of money you can rob tonight without alerting the police.


```python
from typing import List
```

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        n = len(nums)

        # we make a solve function, this function makes a choice of whether we wanna pick up the house and rob it or if we wanna rob the next house
        # we care about solving from i and onwards
        def solve(i):

            # Step 1: we define a base case, if i is out of bounds then we return 0
            if i >= len(nums) or i<0:
                return 0

            # step 2; if that is not the case at every i we have a choice
            money_made_from_robbing_i = nums[i] + solve(i+2)
            money_made_from_skipping_i = solve(i+1)      

            return max(money_made_from_robbing_i, money_made_from_skipping_i)

        # now we need to know how much a robber would make by the time he robs the last house
        return solve(0)
```

### Memoization

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        memo = {}
        def solve(i):
            if i<0 or i>=len(nums):
                return 0
            if i in memo:
                return memo[i]
            pick_i = nums[i] + solve(i+2)
            skip_i = solve(i+1)
            res = max(pick_i, skip_i)
            memo[i] = res
            return res
        return solve(0)
```

### Tabulation

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        
        n = len(nums)
        dp = [0]*(n+2) # we add 2 extra and initialize them to 0

        dp[n+1] = 0
        dp[n] = 0

        

        for i in range(n-1, -1, -1):
            # choose to rob i
            pick_i = nums[i] + dp[i+2]
            # choose to skip i
            skip_i = dp[i+1]
            dp[i] = max(pick_i, skip_i)

        return dp[0]
```

### Space Optimization


```python
class Solution:
    def rob(self, nums: list[int]) -> int:

        n = len(nums)

        a = 0
        b = 0

        for i in range(n-1, -1, -1):
            ith_value = max(nums[i]+b, a)
            b = a
            a = ith_value

        return a
```
