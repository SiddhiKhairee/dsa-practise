### Subsets

Given an integer array nums of unique elements, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.

 

Example 1:

Input: nums = [1,2,3]
Output: [[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]
Example 2:

Input: nums = [0]
Output: [[],[0]]

```python
path = [1, 2]
output = []

output.append(path)

print(output)  # [[1, 2]]
path.pop()

# everytime we modify path, we also end up modifying the output
print(path)    # [1]
print(output)  # [[1]]
```

### For loop backtracking solution

```python
class Solution:
    def subsets(self, nums: list[int]) -> list[list[int]]:

        # we make an output array call subsets
        subsets = []

        # we then make a dfs function
        # this function will take a start_index and a path
        def dfs(start_index, path):

            # we dont need a base case here as we are using a for loop
            # if we didnt use one, our base case would have been
            # if start_index == len(nums) return

            # we start off this function, by simply appending whatever path there is
            # we use copy, so that changing the path would not change the output
            subsets.append(path.copy())

            #we define a for loop,
            # this loop gos from start index of 0 to len(n)
            for i in range(start_index, len(nums)):

                # CHOOSE: we append the element to the path
                path.append(nums[i)]
                # EXPLORE: we call the dfs function on the next element
                dfs(i+1, path)
                # UNDO: whenver this function executes and return something, we pop one element from the path so we can explore other choices
                path.pop()

        dfs(0, [])
        return subsets
```

### Backtracking solution

```python
class Solution:
    def subsets(self, nums: list[int]) -> list[list[int]]:

        subsets = []
        # define a dfs function
        def dfs(start_index, path):

            if start_index == len(nums):
                # add the path to the output
                subsets.append(path.copy())
                return 

            # EXPLORE by excluding current
            dfs(start_index+1, path)

            # EXPLORE by including current element
            path.append(nums[start_index])
            dfs(start_index+1, path)

            # UNDO
            path.pop()

        dfs(0, [])
        return subsets
            

    
```

### Time and Space Complexity

Time Complexity: \(O(n \cdot 2^n)\)



Auxiliary Space Complexity: \(O(n)\)



Total Space Complexity (including output): \(O(n \cdot 2^n)\)

My algorithm uses backtracking to generate all possible subsets. Since each of the \(n\) elements can either be included or excluded, there are \(2^n\) subsets. I create a copy of each subset, which takes up to \(O(n)\) time. Therefore, the overall time complexity is \(O(n \cdot 2^n)\). The recursion stack and current path each take at most \(O(n)\) space, giving \(O(n)\) auxiliary space. Including the output, the total space complexity is \(O(n \cdot 2^n)\).
