### Path Sum

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

Your function has this responsibility:
dfs(node, remaining_sum) determines whether a root-to-leaf path starting at node can contribute exactly remaining_sum.

That means:
- When you arrive at a node, its value has not yet been subtracted.
- If the node is a leaf, compare its value with remaining_sum.
- If the node has children, subtract its value before exploring them.

```python
class Solution:
    def hasPathSum(self, root: TreeNode | None, targetSum: int) -> bool:
        remaining_sum = targetSum
        def dfs(root, remaining_sum):
            # base case
            if root is None:
                return False

            # our second base case will be if root exists and it is a leaf_node, then we check if remaining sum == 0, if it isnt it means we return false
            if root.left is None and root.right is None:
                # check if remaining_sum is equal to the current node value:
                if remaining_sum == root.val:
                    return True

            # if our root exists and if the root has children then
            left_subtree = dfs(root.left, (remaining_sum - root.val))
            right_subtree = dfs(root.right, (remaining_sum - root.val))

            return left_subtree or right_subtree

        return dfs(root, remaining_sum)
```

Time and space complexity
- Time: O(n) — In the worst case, we visit every node in the tree.
- Auxiliary space: O(h) — The recursion stack depends on the tree's height.
  - Balanced tree: O(log n)
  - Skewed tree: O(n)
