### Factorial using recursion

```python
from typing import List
```

```python
# using return value recursion
def factorial(n):

    if n == 0:
        return 1
    else:
        return (n*factorial(n-1))
```

```python
# using accumalator recursion
def factorial(n, output=1):
    if n==0:
        return output
    return factorial(n-1, n*output)
```
