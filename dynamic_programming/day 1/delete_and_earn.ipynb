### Delete and Earn

You are given an integer array nums. You want to maximize the number of points you get by performing the following operation any number of times:

Pick any nums[i] and delete it to earn nums[i] points. Afterwards, you must delete every element equal to nums[i] - 1 and every element equal to nums[i] + 1.
Return the maximum number of points you can earn by applying the above operation some number of times.

 

Example 1:

Input: nums = [3,4,2]
Output: 6
Explanation: You can perform the following operations:
- Delete 4 to earn 4 points. Consequently, 3 is also deleted. nums = [2].
- Delete 2 to earn 2 points. nums = [].
You earn a total of 6 points.
Example 2:

Input: nums = [2,2,3,3,3,4]
Output: 9
Explanation: You can perform the following operations:
- Delete a 3 to earn 3 points. All 2's and 4's are also deleted. nums = [3,3].
- Delete a 3 again to earn 3 points. nums = [3].
- Delete a 3 once more to earn 3 points. nums = [].
You earn a total of 9 points.

### Using Recursion

```python
class Solution:
    def deleteAndEarn(self, nums: list[int]) -> int:

        # make a hashmap of nums and its frequency
        freq_hsmp = {}
        for num in nums:
            if num in freq_hsmp:
                freq_hsmp[num] += 1
            else:
                freq_hsmp[num] = 1

        # create an array, which is sorted
        arr = sorted([k for k in freq_hsmp.keys()])

        # solve i will return all the points that have been collected so far by deleting the ith element, will consider all the cells before i 
        def solve(i):

            if i<0 or i>=len(arr):
                return 0

            value_i = arr[i]*freq_hsmp[arr[i]]

            # we choose arr[i], this means we gotta skip solve(i-1) provided arr[i-1] == arr[i] - 1
            if i!=0 and arr[i-1] == arr[i] - 1:
                choose_i = value_i + solve(i-2)
            else:
                choose_i = value_i + solve(i-1)

            # we skip arr[i], this means we choose solve(i-1)
            skip_i = solve(i-1)

            return max(choose_i, skip_i)
        return solve(len(arr)-1)
```

### Memoization

```python
from functools import cache
class Solution:
    def deleteAndEarn(self, nums: list[int]) -> int:

        # frequency hashmap
        freq_hsmp = {}
        for num in nums:
            if num in freq_hsmp:
                freq_hsmp[num] += 1
            else:
                freq_hsmp[num] = 1

        # sorted arr with no duplicates
        arr = sorted([k for k in freq_hsmp.keys()])

        @cache
        def solve(i):

            if i<0 or i>=len(arr):
                return 0

            value_i = arr[i]*freq_hsmp[arr[i]]

            # choose arr[i]
            if i!=0 and arr[i-1] == arr[i] - 1:
                choose_i = value_i + solve(i-2)
            else:
                choose_i = value_i + solve(i-1)

            # skip_i
            skip_i = solve(i-1)

            return max(choose_i, skip_i)

        return solve(len(arr)-1)

```

### Dynamic Programming Tabulation approach

```python
class Solution:
    def deleteAndEarn(self, nums: list[int]) -> int:
        # frequency hashmap
        freq_hsmp = {}
        for num in nums:
            if num in freq_hsmp:
                freq_hsmp[num] += 1
            else:
                freq_hsmp[num] = 1

        # sorted arr with no duplicates
        arr = sorted([k for k in freq_hsmp.keys()])

        if len(arr) == 1:
            return arr[0]*freq_hsmp[arr[0]]

        # create a dp
        dp = [0]*(len(arr))

        dp[0] = arr[0]*freq_hsmp[arr[0]]

        # for dp[1], we have 2 options pick 1 or skip 1, and dp[1] will have max of the 2 options
        pick_1 = arr[1]*freq_hsmp[arr[1]]

        if (arr[1] - 1) != arr[0]:
            pick_1 += dp[0]

        skip_1 = dp[0]

        dp[1] = max(pick_1, skip_1)

        for i in range(2, len(arr)):

            if arr[i]-1 == arr[i-1]:
                pick_i = arr[i]*freq_hsmp[arr[i]] + dp[i-2]
            else:
                pick_i =  arr[i]*freq_hsmp[arr[i]] + dp[i-1]

            skip_i = dp[i-1]
            dp[i] = max(pick_i, skip_i)


        return dp[-1]      
```
