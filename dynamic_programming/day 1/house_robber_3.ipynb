### House Robber - III

The thief has found himself a new place for his thievery again. There is only one entrance to this area, called root.

Besides the root, each house has one and only one parent house. After a tour, the smart thief realized that all houses in this place form a binary tree. It will automatically contact the police if two directly-linked houses were broken into on the same night.

Given the root of the binary tree, return the maximum amount of money the thief can rob without alerting the police.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def rob(self, root: TreeNode | None) -> int:

        # dfs function will return
        # 1. money obtained from the subtree, by robbing current node
        # 2. money obtained from the subtree by skipping current node
        def dfs(node):

            # if node doesn't exist return None
            if node is None:
                return (0,0)

            # for every node we need to know whether we gonna skip it or pick it
            # say we already have calculate the values for left and right subtrees of the node (cause based on those we need to pick or skip)

            left_rob, left_skip = dfs(node.left)
            right_rob, right_skip = dfs(node.right)

            # we will rob a node which means we include its values, but skip left and right
            pick_node = node.val + left_skip + right_skip

            # if we skip the node, we can rob any child left of right
            # but for that left child and right child, we can still make a choice of whether we wanna pick it or skip it
            # so the way we make that choice is independently, for left -> pick/skip choose max; for right -> pick/skip choose max, and then add it up for skip the node

            skip_node = max(left_skip, left_rob) + max(right_skip, right_rob)

            return (pick_node, skip_node)

        return max(dfs(root)) 
```
