### Maximum Product Subarray

Given an integer array nums, find a subarray that has the largest product, and return the product.

The test cases are generated so that the answer will fit in a 32-bit integer.

Note that the product of an array with a single element is the value of that element.

 

​​​​​​​Example 1:

Input:
 nums = [2,3,-2,4]
Output: 6
Explanation: [2,3] has the largest product 6.
Example 2:

Input: nums = [-2,0,-1]
Output: 0
Explanation: The result cannot be 2, because [-2,-1] is not a subarray.

### Incomplete solution

we needed to check two states this time, because inlike simplw addition, two negetive numbers on its own can have a large positive product

-10 * -5 = 50

we have to thus keep a track of max_dp: whcih is the maximum product seen so far and min_dp with is the minum product seen so far

```python
class Solution:
    def maxProduct(self, nums: list[int]) -> int:

        # we use dynamic programming for this
        # we create a dp array, the first element of the array will be initialized by nums[i]
        # after that every dp[i] will be a chice between max of dp[i-1]*nums[i], nums[i]

        dp = [nums[0]]*len(nums)
        dp[0] = nums[0]
        for i in range(1, len(nums)):
            dp[i] = max(dp[i-1]*nums[i], nums[i])

        return max(dp)
```

### Correct approach

```python
class Solution:
    def maxProduct(self, nums: List[int]) -> int:

        # we use dynamic programming
        # we have to make 2 different dp arrays
        # max_dp will hold the maximum subproduct upto ith element
        # min_dp will hold the minimum subproduct upto the ith element
        # the reason why we are including minimum product is because we never know when we get a -ve number and we multpily it with another -ve number and we end up getting max product of a subarray so far


        max_dp = [nums[0]] * len(nums)
        min_dp = [nums[0]] * len(nums)

        for i in range(1, len(nums)):

            # we first update the max_dp, we consider max of all these options
            # option 1: start fresh -> include just nums[i]
            # option 2: we consider min_dp[i-1]*nums[i]
            # option 3: we consider max_dp[i-1]*nums[i]

            max_dp[i] = max(nums[i], min_dp[i-1]*nums[i], max_dp[i-1]*nums[i])

            # we update the min_dp in the same way, we consider the same 3 options, but this time we choose the min one
            min_dp[i] = min(nums[i], min_dp[i-1]*nums[i], max_dp[i-1]*nums[i])


        # by the end of this, we have considered all the possibilities, and we can return max(max_dp)
        return max(max_dp)
```

### Space Optimization

```python
class Solution:
    def maxProduct(self, nums: list[int]) -> int:

        current_max_prod = nums[0]
        current_min_prod = nums[0]

        global_max = nums[0]

        for i in range(1, len(nums)):
            new_max = max(nums[i], current_max_prod*nums[i], current_min_prod*nums[i])
            new_min = min(nums[i], current_max_prod*nums[i], current_min_prod*nums[i])
            current_max_prod, current_min_prod = new_max , new_min
            global_max = max(global_max, current_max_prod)

        return global_max
        
```
