### Maximum Subarray

Given an integer array nums, find the subarray with the largest sum, and return its sum.

 

Example 1:

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.
Example 2:

Input: nums = [1]
Output: 1
Explanation: The subarray [1] has the largest sum 1.
Example 3:

Input: nums = [5,4,-1,7,8]
Output: 23
Explanation: The subarray [5,4,-1,7,8] has the largest sum 23.

```python
from typing import List
```

```python
class Solution:
    def maxSubArray(self, nums: list[int]) -> int:

        # we use a dynamic programming based solution
        # we have to create a dp array
        # dp[i] will be a choice of dp[i-1]+nums[i] ot just nums[i]. whatever is the maximum
        # we initalize dp
        dp = [0] * len(nums)
        dp[0] = nums[0]
        for i in range(1, len(nums)):
            dp[i] = max(nums[i] , dp[i-1]+nums[i])

        # we return the max in dp. cause that is our maximum subarray
        return max(dp)
```

### Space optimized approach

```python
class Solution:
    def maxSubArray(self, nums: list[int]) -> int:

        # we initialize both of them with the first element
        current_sum = nums[0]
        max_sum = nums[0]

        # we traverse thru rest of the array
        for i in range(1, len(nums)):
            current_sum = max(current_sum+nums[i], nums[i])
            max_sum = max(current_sum, max_sum)

        return max_sum
       
```
