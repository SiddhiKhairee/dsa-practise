### Min Stack (Medium)

Design a stack that supports push, pop, top, and retrieving the minimum element in constant time.

Implement the MinStack class:

MinStack() initializes the stack object.
void push(int value) pushes the element value onto the stack.
void pop() removes the element on the top of the stack.
int top() gets the top element of the stack.
int getMin() retrieves the minimum element in the stack.
You must implement a solution with O(1) time complexity for each function.

 

```python
from typing import List
```

```python
class MinStack:

    def __init__(self):
        self.arr = []
        self.mini = []
        

    def push(self, value: int) -> None:
        self.arr.append(value)
        if len(self.mini) == 0:
            self.mini.append(value)
        else:
            # check what element is in the mini, if the element in the mini, is greater than the value, we pop the mini element and replace it with the value
            self.value_added_to_mini = min(value, self.mini[-1])
            self.mini.append(self.value_added_to_mini)
        return None
        

    def pop(self) -> None:
        self.arr.pop()
        self.mini.pop()

        return None
        

    def top(self) -> int:
        if self.arr:
            return self.arr[-1]
        

    def getMin(self) -> int:
        if self.mini:
            return self.mini[-1]
```
