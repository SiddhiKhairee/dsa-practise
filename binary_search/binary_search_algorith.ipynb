### Binary Search Algorithm

Time Complexity: O(log n)
Works on a sorted array

```python
from typing import Lists
```

```python
class Solution:
    def binary_search(self, arr: list[int], target: int) -> int:
        left = 0
        right = len(arr) - 1

        while left <= right:
            middle_index = (left + right) // 2
            # if middle element is target
            if arr[middle_index] == target:
                return middle_index
            
            # if middle element is bigger than target, means our target will be in the lefy half
            elif arr[middle_index] > target:
                right = middle_index - 1
            
            # if middle element is smaller than target, means our target is in the right half
            elif arr[middle_index] < target:
                left = middle_index + 1

        return -1

    
```

```python
s = Solution()
result = s.binary_search([1,2,3,4,5], 0)
print(result)
```
