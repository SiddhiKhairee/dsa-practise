### Merge two sorted lists (easy)
You are given the heads of two sorted linked lists list1 and list2.

Merge the two lists into one sorted list. The list should be made by splicing together the nodes of the first two lists.

Return the head of the merged linked list
Input: list1 = [1,2,4], list2 = [1,3,4]
Output: [1,1,2,3,4,4]
Example 2:

Input: list1 = [], list2 = []
Output: []
Example 3:

Input: list1 = [], list2 = [0]
Output: [0]
 

```python
from typing import Optional
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:

        # make a dummy node to handle the edge case where head needs to be changed
        dummy = ListNode()
        tail = dummy

        # compare the values in the two list

        # we do this until any one list is shorter and runs out of values
        while list1 and list2:

            # case 1: list1 has a smaller value, we add that value to our dummy tail node, and move l1 foward
            if list1.val < list2.val:
                tail.next = list1
                list1 = list1.next

            # case 2: list2 has a smaller value, we add that value to our dummy tail node, and move l2 foward
            else:
                tail.next = list2
                list2 = list2.next

            # regardless of whatever the case is, tail will always move forward
            tail = tail.next

        # say one of the list is shorter, and the values are done, then we simply take the longer list and append it as it is to the tail

        if list1:
            tail.next = list1
        elif list2:
            tail.next = list2

        # return dummy head, dummy currently is a prev Null node 
        return dummy.next
```
