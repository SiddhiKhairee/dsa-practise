### Palindrome Linked List (Easy)
Given the head of a singly linked list, return true if it is a palindrome or false otherwise.

 

Example 1:


Input: head = [1,2,2,1]
Output: true
Example 2:


Input: head = [1,2]
Output: false
 

Constraints:

The number of nodes in the list is in the range [1, 105].
0 <= Node.val <= 9

```python
from typing import Optional
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:

        # Step 1: get slow to the middle of the list
        slow = head
        fast = head
        before_slow = head
        while fast and fast.next:
            fast = fast.next.next
            before_slow = slow
            slow = slow.next

        before_slow.next = None
            
            
    


        # Now from this middle we reverse the list
        prev = None
        curr2 = slow
        while curr2:
            temp_next = curr2.next
            curr2.next = prev
            prev = curr2
            curr2 = temp_next

        # compare prev to curr1
        curr1 = head
        while curr1:
            if curr1.val!=prev.val:
                return False
            curr1 = curr1.next
            prev = prev.next

        return True

```
