### Reverse a Linked List (Easy)
Given the head of a singly linked list, reverse the list, and return the reversed list
Input: head = [1,2,3,4,5]
Output: [5,4,3,2,1]
Input: head = [1,2]
Output: [2,1]
Example 3:

Input: head = []
Output: []
 

Constraints:

The number of nodes in the list is the range [0, 5000].
-5000 <= Node.val <= 5000

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
from typing import Optional


class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:

        # we define a current node as head
        curr = head
        # we define a previous node, which will be before head (our current node)
        prev = None

        #simply for traversing a ll, we use the following loop
        while curr:
            # currently head is pointing to head.next, we need to store this next value (aka node 2 value somewhere)
            temp_next = curr.next 
            # now we need to ensure that the current node will point to prev node
            curr.next = prev
            # we need to update the prev and curr values
            # the prev node will move one node furtur, so whatever is our curr value, becomes our new prev
            prev = curr
            # to change curr value, we will use the temparat variable where we had stored the next value 
            curr = temp_next

        # by the end of it, our prev will have reached the last node, and now this node will be the new head
        return prev
```
