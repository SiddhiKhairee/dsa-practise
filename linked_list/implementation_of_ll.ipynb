### Implementing a Linked List in Python

```python
from typing import List
```

Node represents an individual element in the Linked List. 
This class will have 2 class members
1. Data: Can contain number, integers or complex objects
2. Next which is the pointer to the next element

```python
# create a class node
class Node:
    def __init__(self, data=None, next=None):
        self.data = data
        self.next = next
        
```

We create another class called Linked List. This class has a member called head which will point to the start of the Linked List

```python
# create another class linked list 
class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_begining(self, data):
        # we want to insert an element to the start of the list. So we make a new node
        # this is done by creating an object of the node class
        # since we are inserting at begining, the current head will be considered as "next" pointer to this new node
        node = Node(data, self.head)
        # once node is created, we change the head value to the newly created node
        self.head = node

    def print(self):
        # say if we have no value at head, we print that the list is empty
        if self.head is None:
            print("Linked List is empty")
            return

        #else we set a variable i as an iterator
        #initial value of iterator will be the head,
        # while this head exists we update the iterator value by assigning it the next element
        i = self.head
        ll_string = ''
        while i:
            ll_string += str(i.data) + '-->'
            i = i.next
        print(ll_string)

    # we define another method for inserting at the end of the linked list
    def append(self, data):

        # Check if head is empty, if it is, your new data becomes the head
        if self.head is None:
            self.head = Node(data, None)
            return

        # if linked list is not empty, what we wanna do is iterate to the end of linked list
        # once it reaches the end, we append a new element
        i = self.head
        while i.next: # if i.next has some value it means Our iterator is not in the end
            # this means that we update the iterator value
            i = i.next

        # when i.next becomes null, it means that we are at our last element
        # now i.next will be our new inserted node
        i.next = Node(data, None)

    # take multiple values from list, and make them into a linked list
    def insert_values(self, data_list):
        self.head = None 
        # we need to make sure that the head is clear
        for data in data_list:
            self.append(data)

    # implement a function to get the length of linked list
    def get_length(self):
        c = 0
        i = self.head
        while i:
            c += 1
            i = i.next
        return c

    # remove an element from a particular index
    def remove_at(self, index):
        if index<0 or index>=self.get_length():
            raise Exception("Index is out of bounds")

        elif index == 0:
            # this means we are removing the head element, so the next element becomes our head
            self.head = self.head.next
            return
        # in other cases, we wanna remove an element at index = 3
        # so we set up a count variable,
        # when this count reaches 2, it means that our i is reached 2nd positon,
        # here i.next is pointing at 3, so we do i.next.next so it points at 4 instead
        # so we just from index = 2 to index = 4 according to our iterator, thereby removing 3
        else:
            c = 0
            i = self.head
            while i:
                if c == index - 1:
                    i.next = i.next.next
                    break
                i = i.next
                c += 1

    # our next function is inserting into a particular index
    def insert_at(self, index, data):
        if index<0 or index>=self.get_length():
            raise Exception("Index is out of bounds")

        elif index == 0:
            self.insert_at_begining(data)
            return

        else:
            c = 0
            i = self.head
            while i:
                if c == index - 1:
                    # create a new node
                    node = Node(data, i.next)
                    i.next = node
                    break

                i = i.next
                c += 1


    
        






if __name__ == '__main__':
    ll = LinkedList()
    ll.insert_values(['apple', 'mango', 'banana', 'grape'])
    ll.print()
    ll.remove_at(1)
    ll.insert_at(0, 'peaches')
    print(ll.get_length())
    ll.print()


```
