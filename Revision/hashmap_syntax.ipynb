### Hashmap Syntax

```python
from typing import List
```

```python
# convert an integer array into a hashmap that stores the array elements and their frequencies
def arr_to_hsmp(arr: list[int]) -> dict:

    #create an empty dictionary
    hsmp = {}
    
    for i in arr:
        # if key already exists in hsmp, we simply update the value
        if i in hsmp:
            hsmp[i] += 1
        # if key not in hsmp, we add the key to the hashmap, and assign the value 1 to it
        else:
            hsmp[i] = 1

    return hsmp


```

```python
print(arr_to_hsmp([1,2,3,4,5,6,7,1,2,2,3,4,5,9,1,2]))
```

```python
# various functions in the hashmap
result = arr_to_hsmp([1,2,3,4,5,6,7,1,2,2,3,4,5,9,1,2])
print(result.get(1)) # will return the value
print(result.get(22, 0)) # will return None if no argument, 0 if argument
```

```python
del result[1] # deleted the key 1
print(result)
```

```python
del result[22] # since key missing we get an error
print(result)
```

```python
result.pop(22) # will also give key error if none is not mentioned
print(result)
```

```python
result.pop(22, None) # will also give key error if none is not mentioned
print(result)
```

```python
# convert an integer array into a hashmap that stores the array elements and their frequencies
def arr_to_hsmp(arr: list[int]) -> dict:

    #create an empty dictionary
    hsmp = {}
    
    for i in arr:
        # check if hsmp has the key
        val = hsmp.get(i, 0)
        hsmp[i] = val + 1

    return hsmp


```

```python
from collections import Counter
def arr_to_hsmp(arr: list[int]) -> dict:

    # we simply give counter the array
    counts = Counter(arr)
    

    return dict(counts)
```

```python
def arr_to_hsmp(arr: list[int]) -> dict:

    hsmp = defaultdict(int)
    for i in arr:
        hsmp[i] += 1
    return hsmp
```
