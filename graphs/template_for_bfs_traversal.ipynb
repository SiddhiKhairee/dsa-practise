```python
def bfs(root):
    # if root exists, we enqueue it
    queue = deque([root])
    # we make a set of all the visited nodes
    visited = set([root])

    while queue:
        # we pop out of a queue
        node = queue.popleft()
        for neighbors in get_neighbors(node):
            if neighbors in visited:
                continue
            queue.append(neighbors)
            visited.add(neighbors)
```

```python
def bfs_with_levels(root):
    # starting level will be 0
    level = 0
    queue = deque([root])
    visited = set([root])

    while len(queue) > 0:
        n = len(queue) # will give number of nodes on each level
        for i in range(n):
            # dequeue the node
            node = queue.popleft()
            for neighbor in get_neighbors(node):
                if neighbor in visited:
                    continue
                else:
                    # add the neighbor to the queue
                    # add the neighbor to visited
                    queue.append(neighbor)
                    visited.add(neighbor)
        # at end of each level, we add one to level traversed
        level += 1
```
