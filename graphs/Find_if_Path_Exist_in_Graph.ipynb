### Find if Path Exists in Graph

There is a bi-directional graph with n vertices, where each vertex is labeled from 0 to n - 1 (inclusive). The edges in the graph are represented as a 2D integer array edges, where each edges[i] = [ui, vi] denotes a bi-directional edge between vertex ui and vertex vi. Every vertex pair is connected by at most one edge, and no vertex has an edge to itself.

You want to determine if there is a valid path that exists from vertex source to vertex destination.

Given edges and the integers n, source, and destination, return true if there is a valid path from source to destination, or false otherwise.

```python
from collections import deque
class Solution:
    def validPath(self, n: int, edges: list[list[int]], source: int, destination: int) -> bool:
        adjacency_list = [[] for i in range(n)]

        for edge in edges:
            u = edge[0]
            v = edge[1]

            if v not in adjacency_list[u]:
                adjacency_list[u].append(v) # add v to the uth index
            
            if u not in adjacency_list[v]:
                adjacency_list[v].append(u)

        # now that we have an adjacency list, we will simply habe to find a path from the source node to the destination node
        # we can do that by a simple bfs traversal

        if source == destination: # we are on the same node
            return True

        # add the source node to the queue
        queue = deque([source])
        # we also add this node to visited_nodes set
        visited_nodes = set()
        visited_nodes.add(source)
        # we start with our enqueuing and dequeuing operations
        while len(queue) > 0:
            # let n be the number of nodes in a queue that are present on the same level
            level_size = len(queue)
            # for every level, we dequeue a node
            for i in range(level_size):
                node = queue.popleft()
                # if this node is the destination node, we return traversed edges
                if node == destination:
                    return True
                # if thats not the case, we simply add the neighbors of the node to the queue, provided they have not been visited before
                # we check for the neighbors of the nodes
                for neighbor in adjacency_list[node]:
                    if neighbor in visited_nodes:
                        continue
                    else:
                        # we add the neighbors to the queue and to the visited_nodes 
                        queue.append(neighbor)
                        visited_nodes.add(neighbor)

        return False



```

### Time and Space Complexity

Time: O(n + E). Building the adjacency list is O(E), and BFS visits each vertex once and looks at each edge twice (once from each end).
Space: O(n + E) for the adjacency list, plus O(n) for visited and the queue.
