### Course Scheduling

There are a total of n courses a student has to take, numbered from 0 to n-1. A course may have prerequisites. The "depends on" relationship is expressed as a pair of numbers. For example, [0, 1] means you need to take course 1 before taking course 0. Given n and the list of prerequisites, decide if it is possible to take all the courses.

Example 1:

Input: n = 2, prerequisites = [[0, 1]]

Output: true

Explanation: we can take 1 first and then 0.

Example 2:

Input: n = 2, prerequisites = [[0, 1], [1, 0]]

Output: false

Explanation: the two courses depend on each other, there is no way to take them

```python
from collections import deque
```

```python
class Solution:
    def canFinish(self, numCourses: int, prerequisites: list[list[int]]) -> bool:

        indegree_hashmap = {}
        for i in range(numCourses):
            indegree_hashmap[i] = 0
        for element in prerequisites:
            indegree_hashmap[element[1]] += 1
        
        # neighbors hashmap: will contain all the neighbors of the previous node
        neighbors_hashmap = {}
        for i in range(numCourses):
            neighbors_hashmap[i] = []
        for element in prerequisites:
            neighbors_hashmap[element[0]].append(element[1])
        
        queue = deque()
        visited = set()
        processed_nodes = []

        for node in indegree_hashmap:
            if indegree_hashmap[node] == 0:
                queue.append(node)
                visited.add(node)

        while queue:
            que_len = len(queue)
            for _ in range(que_len):
                popped_node = queue.popleft()
                processed_nodes.append(popped_node)

                neighbors_of_popped_node = neighbors_hashmap[popped_node]

                for neighbor in neighbors_of_popped_node:
                    
                        indegree_hashmap[neighbor] -= 1
                        if neighbor not in visited and indegree_hashmap[neighbor] == 0:
                            queue.append(neighbor)
                            visited.add(neighbor)

        return len(processed_nodes) == numCourses
```
