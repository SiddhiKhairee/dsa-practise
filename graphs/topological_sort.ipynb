### Topological Sort

### Kahn's Algorithm



Let's summarize the steps here:

Initialize a hashmap to store the in-degrees.

Go through the nodes, count the in-degree of each node.

Push the nodes with 0 in-degree into the queue.

Pop each node from the queue, subtract 1 from the in-degree of each of its neighbors (each node it points to).

If a node's in-degree drops to 0, then push it into the queue.
repeat until the queue is empty. If any nodes remain unprocessed, then there must be a cycle.

```python
from collections import deque
```

```python
# we assume we have a graph represented as an adjacency list 
graph = [[], [], []] # ith element represents all the nodes that are being pointed to by outedges of i
```

```python
def topological_sort(graph):

    # lets create an indegree function
    # this function will take the graph
    # will return a hashmap where every node will correspond to the number of inedges
    def indegrees(graph):
        indegree_hsmp = {}

        for node in graph:
            indegree_hsmp[node] = 0 # we initialize the node as 0

        # currently we have a hashmap which has all the nodes initialized, but all their values (indegrees) are zero
        # to calculate indegree, we go through each element in the graph
        # every time we encounter a neighbor within the element, we know that that neighbor will have an incoming edge from that element
        # and so we update the value of that neighbor in the hashmap
        for node in graph:
            for neighbor in node:
                indegree_hsmp[neighbor] += 1

        # by the end of this loop our hashmap will have both the nodes and all the indegrees of the incoming edges
        return indegree_hsmp


    queue = deque()
    results = [] # this list will store all the nodes in correct topological order
    # for topological sort, we first call the hashmap
    indegree_hashmap = indegrees(graph)
    # once we have the hashmap
    # we perform bfs, we enqueue all the elements which have the indegree aka value 0
    for node in indegree_hashmap:
        if indegree_hashmap[node] == 0: # if the value of any node is 0, we enqueue it
            queue.append(node)

    # for all the nodes that have been appended, we pop them
    # we take its neighbors using the graph adjacency list
    # reduce these neighbors' indegree values by 1
    # if any other indegree value is reaching 0, append it to the queue
    while queue:
        node = queue.popleft()
        # this popped node goes in result
        results.append(node)
        for neighbor in graph[node]:
            indegree_hashmap[neighbor] -= 1 # for the popped node, we will search its neighbors and reduce the indegree by 1
            # now we check again if any neighbor has reached 0
            # if it has we enqueue it so it can be popped later
            if indegree_hashmap[neighbor] == 0: # Note: we do not keep a visited set, because, we only append to the queue if indegree is zero
                queue.append(neighbor)          # Note: visited set usually keeps a track of cycle exists or not in traversing, but if cycle exists then indegree would bever reach 0, so we are good

    # by the end of bfs, hopefully
    # all the indegrees have reached 0
    # implies that all nodes have been added atleast once to the queue, and have been popped out of the queue
    # and thus, appended to the result

    if len(results) == len(graph):
        return results
    else:
        return None # there is a cycle

```
