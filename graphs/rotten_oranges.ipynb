### Rotten Oranges

You are given an m x n grid where each cell can have one of three values:

0 representing an empty cell,
1 representing a fresh orange, or
2 representing a rotten orange.
Every minute, any fresh orange that is 4-directionally adjacent to a rotten orange becomes rotten.

Return the minimum number of minutes that must elapse until no cell has a fresh orange. If this is impossible, return -1.

```python
from collections import deque
```

```python
class Solution:
    def orangesRotting(self, grid: list[list[int]]) -> int:

        # def a get neighbors function
        def get_neighbors(r,c,grid):
            rows = len(grid)
            cols = len(grid[0])
            directions = [
                (-1,0),
                (0,1),
                (1,0),
                (0,-1)
            ]
            neighbors = []
            for dr, dc in directions:
                nr = r+dr
                nc = c+dc
                if 0<=nr<rows and 0<=nc<cols:
                    # check if the neighbors are 1, if yes then only output them
                    if grid[nr][nc] == 1:
                        neighbors.append((nr,nc))

            return neighbors

        queue = deque([])
        minutes = 0
        fresh_oranges = 0
        
        # we traverse through the loop once, and update the number of fresh oranges and enqueue all the r,c which already have rotten oranges
        for i in range(len(grid)):
            for j in range(len(grid[0])):
                if grid[i][j] == 1:
                    fresh_oranges += 1
                if grid[i][j] == 2:
                    queue.append((i,j))
                    

        # now we have the total number of fresh oranges
        # we also have a queue which has all the cells with rotten oranges
        
        while len(queue) > 0 and fresh_oranges > 0:
            # we perform a bfs traverse on the grid we dequeue all the elements on the level
            n = len(queue)
            for _ in range(n): 
                r,c = queue.popleft()

                for r, c in get_neighbors(r,c,grid):
                    
                    grid[r][c] = 2
                    # reduce the number of fresh oranges
                    fresh_oranges -= 1
                    queue.append((r,c))
                        

            minutes += 1

        if fresh_oranges != 0:
            return -1
        else:
            return minutes
```

```python
s = Solution()
res = s.orangesRotting([[2,1,1],[1,1,0],[0,1,1]])
print(res)
```

### Time and Space Complexity

Rotting Oranges (BFS):

Time: O(R×C). The initial scan touches every cell once. Each cell goes into the queue at most once, because it flips from 1 to 2 when pushed. Each pop checks 4 neighbors, which is a constant.
Space: O(R×C). That’s the queue. Worst case, every cell starts as 2 and the whole grid is in the queue at once. The in-place trick means you don’t need an extra visited set. The queue is still there.
