### Uniform Cost Search

Uniform Cost Search (UCS) is almost identical to the Dijkstra implementation you already wrote.

We only need to make a few changes.
1. Dijkstra vs. Uniform Cost Search
Your current Dijkstra implementation finds the shortest distances from a source node to all reachable nodes.

Uniform Cost Search finds the shortest distance from a source node to one specific destination.
Dijkstra (your version)	Uniform Cost Search
Takes graph, start	Takes graph, start, target
Continues until heap is empty	Stops when target is popped
Returns distance hashmap	Returns shortest distance to target
Uses min-heap	Uses min-heap
Both require non-negative edge weights.

```python
from math import inf
from heapq import heappop, heappush
```

```python
def uniform_cost_search(graph, start, target):

    # step 1: we make a distance hashmap, include all the nodes and set their distances to infinity, except for the start node
    distance = {}
    for node in graph:
        distance[node] = float('inf')
    distance[start] = 0

    # we make a heap and add the start node to it
    min_heap = [(0, start)]

    while min_heap:
        # we pop out of the heap
        popped_distance, popped_node = heappop(min_heap)

        # for the popped_node, we check if its popped_distance, is greater than the distance in the hashmap
        if popped_distance > distance[popped_node]:
            continue # we skip this node, do not update 
        # else, we know that the popped_distance of the node is smaller than the one mentioned in distance hashmap
        # so we first check if this is the target node, if it is we return it
        if popped_node == target:
            return popped_distance
        # otherwise we continue, for the popped_node we get its neighbors and their weights
        for neighbor, edge_weight in graph[popped_node]:
            # assign a new weight to the neighor
            new_weight = edge_weight + popped_distance 
            # if this new weight is smaller than the one is distance hashmap, we update it
            if new_weight < distance[neighbor]:
                distance[neighbor] = new_weight
                # push into the heap
                heappush(min_heap, (new_weight, neighbor))

    # if this function, doesnt return anything, it means the node is unreachable and so we return inf
    return float('inf')
```
