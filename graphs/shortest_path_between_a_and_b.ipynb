### Shortest Path Between A and B

Given an (unweighted) connected graph, return the length of the shortest path between two nodes A and B, in terms of the number of edges.

Assume there always exists a path between nodes A and B.

```python
from collections import deque
```

```python
def shortest_path(graph: list[list[int]], a: int, b: int) -> int:

    # we first define a get_neighbors function
    # this function will take an ith node and will return all its neigbors in a list
    def get_neighbors(i):
        neighbors = graph[i]
        return neighbors # all neighors if ith node are in a list here

    # we use a breadth first search
    # take the source node and add it to the queue
    queue = deque([a])
    # we also add this node to visited_nodes set
    visited_nodes = set()
    visited_nodes.add(a)
    # keep a track of how many edges we have traversed, every time we explore a new node
    # since we are still at source node, we have traversed 0 edges
    traversed_edges = 0
    # we deal with the edge case that, if a and b happen to be the same node, we return 0
    if a == b:
        return 0

    # we start with our enqueuing and dequeuing operations
    while len(queue) > 0:
        # let n be the number of nodes in a queue that are present on the same level
        n = len(queue)
        # for every level, we dequeue a node
        for i in range(n):
            node = queue.popleft()
            # if this node is the destination node, we return traversed edges
            if node == b:
                return traversed_edges
            # if thats not the case, we simply add the neighbors of the node to the queue, provided they have not been visited before
            for neighbor in get_neighbors(node):
                if neighbor in visited_nodes:
                    continue
                else:
                    # we add the neighbors to the queue and to the visited_nodes 
                    queue.append(neighbor)
                    visited_nodes.add(neighbor)

        # we add to number of traversed edges
        traversed_edges += 1

    return traversed_edges

```
