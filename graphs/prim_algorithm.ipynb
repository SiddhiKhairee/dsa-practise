### Prim's Algorithm 

Start from a node and repeatedly add the cheapest edge that connects the growing tree to a new node.

Connect all the nodes in a weighted, undirected graph with the minimum total edge cost, without creating cycles.

We will use:
- Min-heap: To always pick the cheapest available edge.
- Visited set: To keep track of nodes already included in our MST.
- Total cost: To calculate the sum of the selected edge weights.
Step 1: Initialize everything
1. Create an empty visited set.
2. Create a min_heap.
3. Set total_cost = 0.
4. Pick any starting node.
5. Push (0, starting_node) into the min-heap because it costs nothing to start at that node.
Step 2: Run a loop while the min-heap is not empty
As long as there are elements in the min-heap, keep repeating the following steps.
Step 3: Pop the cheapest node from the min-heap
1. Pop the entry with the smallest edge weight.
2. Store the popped weight and node in two variables.
Remember: The weight represents the cost of connecting that node to our existing MST, not the total distance from the starting node.
Step 4: Check whether the node is already visited
1. If the popped node is already in visited, skip it using continue.
2. Otherwise, proceed to the next step.
Why? We don't want to add the same node twice because that could create a cycle.
Step 5: Add the node to our MST
1. Add the popped node to the visited set.
2. Add the popped edge weight to total_cost.
This means we have now connected one more node to our MST.
Step 6: Explore the neighbors of the current node
For every neighbor of the current node:
1. Get the neighbor and the weight of the edge connecting them.
2. Check whether the neighbor is already in visited.
3. If the neighbor is not visited, push (edge_weight, neighbor) into the min-heap.
Important: We push the individual edge weight, not the cumulative distance.

The min-heap now contains candidate edges for connecting new nodes to our MST.
Step 7: Repeat
Go back to Step 3.

Keep popping the cheapest candidate edge, skipping already visited nodes, and adding new nodes until the min-heap becomes empty.

For a connected graph, all nodes will now be included in the MST.
Step 8: Return the result
Return total_cost.

This gives us the minimum total cost required to connect all the nodes.


```python
from heapq import heappop, heappush
```

```python
from heapq import heappop, heappush
def prim(graph, start):

    # Initialize everything
    min_heap = [(0, start)]
    total_cost = 0
    visited = set()

    while min_heap:

        # pop the node from the heap
        popped_edge_weight , popped_node = heappop(min_heap)
        if popped_node in visited:
            continue
        # if it is not in the visited set, then we first add to the visited set and then update the total cost so far
        visited.add(popped_node)
        total_cost += popped_edge_weight

        # we explore the neighbors of the popped_node
        for neighbor, weight in graph[popped_node]:
            # check if neighbor is a part of the visited set
            if neighbor in visited:
                continue
            # if the neighbor is not in visited set, we simply push it to the heap along with its weight
            heappush(min_heap, (weight, neighbor))
    
    # return total cost of the minimum spanning tree
    # we address the edge case where one of the node is completely disconnected
    if len(visited) != len(graph):
        return -1
    return total_cost
```
