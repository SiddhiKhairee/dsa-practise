### Shortest Path First Algorithm (SFPA)

SPFA is actually very similar to the BFS you've already been practicing. The main difference is how we decide whether to add a node to the queue again.

In normal BFS, we usually use a visited set.

In SPFA, we use a distance hashmap to track the shortest distance found so far.

SPFA Algorithm — Simple Steps
1. Create a distance hashmap.
   - Set the source node's distance to 0.
   - Set every other node's distance to infinity (inf).
2. Create a queue.
   - Add the source node to the queue.
3. Start BFS traversal.
   - While the queue is not empty, pop a node from the front.
4. Visit the popped node's neighbors.
   - For each neighbor, calculate:
   new_distance = distance[popped_node] + edge_weight
5. Check if we found a shorter path.
   - If new_distance < distance[neighbor]:
     - Update distance[neighbor] = new_distance.
     - Add the neighbor to the queue if it is not already in the queue.
6. Repeat until the queue is empty.
   - A node can be visited multiple times if a shorter path is discovered.
7. Return the distance hashmap.
   - It contains the shortest distance from the source to every reachable node.

```python
from collections import deque
from math import inf
```

```python
from collections import deque
from math import inf

def shortest_path(graph, start) -> dict:
    def get_neighbors(node: int):
        return graph[node]

    # we make a distance hashmap
    distance_hashmap = {}
    for node in graph:
        distance_hashmap[node] = float('inf')
    distance_hashmap[start] = 0 # the source element must be 0

    # initialize the queue with the source node
    queue = deque([start])
    in_queue = {start} # a set to see which elements are currently in the queue

    while queue:
        popped_node = queue.popleft()
        in_queue.remove(popped_node)

        for neighbor,weight in get_neighbors(popped_node):
            new_distance = distance_hashmap[popped_node] + weight
            if new_distance < distance_hashmap[neighbor]:
                # update the distance hashmap
                distance_hashmap[neighbor] = new_distance
                # add it to the queue but before that we check if it is already in the queue for being processed or not
                if neighbor not in in_queue:
                    queue.append(neighbor)
                    in_queue.add(neighbor)

    return distance_hashmap

```
