### Walls and Gates

```python
from typing import List
```

Input
dungeon_map — a matrix of integer representing the dungeon map.
Output
A matrix of integer representing the dungeon map with the addition of distance to an exit for each empty space.

Example
Input
dungeon_map = [
  [INF, -1, 0, INF],
  [INF, INF, INF, -1],
  [INF, -1, INF, -1],
  [0, -1, INF, INF],
]
Output
[
  [3, -1, 0, 1],
  [2, 2, 1, -1],
  [1, -1, 2, -1],
  [0, -1, 3, 4],
]

```python
from collections import deque
def map_gate_distances(dungeon_map: list[list[int]]) -> list[list[int]]:
    # WRITE YOUR BRILLIANT CODE HERE

    # define a get_neighbors function
    def get_neighbors(dungeon_map, r,c):
        rows = len(dungeon_map)
        cols = len(dungeon_map[0])
        directions = [
            (-1,0),
            (0, 1),
            (1, 0),
            (0, -1)
        ]
        neighbors = []
        for dr, dc in directions:
            nr = r + dr
            nc = c + dc

            if 0<=nr<rows and 0<=nc<cols:
                if dungeon_map[nr][nc] != -1:
                    # if the neighbor is the wall, then we do not append it
                    neighbors.append((nr,nc))
        return neighbors

    def bfs(dungeon_map, start_r, start_c):
        queue = deque([(start_r, start_c)])
        visited = {(start_r, start_c)}
        steps = 0

        while queue:
            n = len(queue)
            for  _ in range(n):
                r,c = queue.popleft()
                if dungeon_map[r][c] == 0: # whenever we reach the gate we return the number of steps taken to reach the gate from a start cell
                    return steps
                for r,c in get_neighbors(dungeon_map, r,c):
                    if (r,c) not in visited:
                        queue.append((r,c))
                        visited.add((r,c))
            steps += 1

        # say we are unable to reach the exit gate, 
        # then we return the actual value of the node instead of number of steps
        return dungeon_map[start_r][start_c]

    for i in range(len(dungeon_map)):
        for j in range(len(dungeon_map[0])):
            if dungeon_map[i][j] != -1 and dungeon_map[i][j] != 0:
                # call bfs on that particular cell of the dungeon map
                steps = bfs(dungeon_map, i, j)
                dungeon_map[i][j] = steps



    return dungeon_map
```
