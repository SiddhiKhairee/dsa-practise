### Flood Fill

In computer graphics, an uncompressed raster image is presented as a matrix of numbers. Each entry of the matrix represents the color of a pixel. A flood fill algorithm takes a coordinate r, c and a replacement color, and replaces all pixels connected to r, c that have the same color (i.e., the pixels connected to the given coordinate with same color and all the other pixels connected to the those pixels of the same color) with the replacement color. (e.g. MS-Paint's paint bucket tool).

You are given an image represented by an m x n grid of integers image, where image[i][j] represents the pixel value of the image. You are also given three integers sr, sc, and color. Your task is to perform a flood fill on the image starting from the pixel image[sr][sc].

To perform a flood fill:

Begin with the starting pixel and change its color to color.
Perform the same process for each pixel that is directly adjacent (pixels that share a side with the original pixel, either horizontally or vertically) and shares the same color as the starting pixel.
Keep repeating this process by checking neighboring pixels of the updated pixels and modifying their color if it matches the original color of the starting pixel.
The process stops when there are no more adjacent pixels of the original color to update.
Return the modified image after performing the flood fill.

 

```python
from typing import List
```

```python
from collections import deque
class Solution:
    def floodFill(self, image: list[list[int]], sr: int, sc: int, color: int) -> list[list[int]]:

        if not image:
            return None

        # we start off by defining a get neighbors function

        def get_neighbors(sr: int, sc:int, image:list[list[int]]) -> List:
            rows_in_image = len(image)
            cols_in_image = len(image[0])
            neighbors = []
            # for every cell in the matrix there are 4 possible neighbors 
            # provided the diagonals aren't considered
            directions = [
                (-1, 0), # upper neighbor
                (0, 1), # right neighbor
                (1, 0), # bottom neigbor
                (0, -1 ) # left neighbor
            ]

            for dr, dc in directions:
                nr = dr + sr 
                nc = dc + sc
                # nr nc represent the coordinates of the neighbors
                # we only add these coordinates to our list on 2 conditions
                # condition 1: these should be within the index bounds
                # condition 2: only the neighbors with same color as the source pixel will ge added
                if 0<=nr<rows_in_image and 0<=nc<cols_in_image:
                    # check for color
                    if image[sr][sc] == image[nr][nc]:
                        neighbors.append((nr,nc))

            return neighbors # all the neighbors of the cell that have the same color as the cell


        def bfs(sr:int, sc:int, image):
            
            if not image:
                return None

            # provide sr and sc always exist within the image
            # we enqueue them
            queue = deque([(sr, sc)])
            # we make a set visited and add these pixels to the set
            visited = {(sr,sc)}

            while queue:

                r, c = queue.popleft()

                # get the neighbors of this popped cell
                for nr,nc in get_neighbors(r,c, image):
                    if (nr, nc) not in visited:
                        queue.append((nr,nc))
                        visited.add((nr,nc))
            # the bfs function will return all the nodes that are connected to the source node, via a path
            # and they have the same color
            return visited

        # we call the bfs function
        # it will return a set 
        reachable_nodes = bfs (sr, sc, image)
        # for each item in set, we change the color
        for coordinates in reachable_nodes:
            i = coordinates[0]
            j = coordinates[1]

            image[i][j] = color

        # by the end of it, all the reachable nodes are of same color
        # we return the image
        return image
    

            
```
