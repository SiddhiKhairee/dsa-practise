### Dijkstra's Algorithm 

Dijkstra's algorithm — simple steps
1. Create a distance hashmap. Set the source node's distance to 0 and every other node's distance to infinity.
2. Create a min-heap. Add (0, source) to it.
3. While the heap is not empty, pop the node with the smallest distance.
4. Check if the popped distance is outdated. If it is greater than the best distance already stored for that node, skip it.
5. Visit all neighbors of the popped node. Calculate the new distance using the edge weight.
6. If a shorter path is found, update the neighbor's distance and push (new_distance, neighbor) into the min-heap.
7. Repeat until the heap is empty, then return the distance hashmap.

```python
from math import inf
import heapq
```

```python
def dikstra(graph, start):

    # step 1: make a distance hashmap
    distance = {}
    for node in graph:
        distance[node] = float('inf')
    distance[start] = 0

    # initialize a min heap
    min_heap = [(0, start)] # give the start node and the distance

    while min_heap:

        # pop from the min heap
        popped_distance, popped_node = heapq.heappop(min_heap)

        # check if the popped_distance is greater than whatever is mentioned in the distance hashmap
        if popped_distance > distance[popped_node]:
            continue
        # we explore neighbors of the popped_node
        for neighbor, edge_weight in graph[popped_node]:
            # for neighbor we calculate the new distance
            new_distance = popped_distance + edge_weight

            # if new distance is smaller 
            if new_distance < distance[neighbor]:
                # update the distance
                distance[neighbor] = new_distance
                # push to the heap
                heapq.heappush(min_heap, (new_distance, neighbor))

    return distance
```
