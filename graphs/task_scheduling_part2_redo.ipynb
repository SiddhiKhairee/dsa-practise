### Task Scheduling (Part 2)

This problem extends Task Scheduling. Each task now has a duration, where times[i] is how long tasks[i] takes.

You can run any number of tasks in parallel, as long as dependencies are respected. If [a, b] appears in requirements, task a must finish before task b can start.

Return the minimum total time required to complete all tasks.

There is guaranteed to be a solution.

```python
def task_scheduling_2(tasks: list[str], times: list[int], requirements: list[list[str]]) -> int:
    # WRITE YOUR BRILLIANT CODE HERE
    # we need multiple hashmaps

    # indegree hashmap: will calculate how many edges enter in a node
    indegree_hashmap = {}
    for task in tasks:
        indegree_hashmap[task] = 0
    for element in requirements:
        indegree_hashmap[element[1]] += 1
    

    # neighbors hashmap: will have a list of all the neighboring nodes
    neighbors_hashmap = {}
    for task in tasks:
        neighbors_hashmap[task] = []
    for element in requirements:
        neighbors_hashmap[element[0]].append(element[1])


    # time hashmap: will have the completion time of each node
    times_hashmap = {}
    for i in range(len(tasks)):
        times_hashmap[tasks[i]] = times[i]


    # finishing time hashmap: will initially have the duration, later be updated during bfs
    finishing_times_hashmap = {}
    for i in range(len(tasks)):
        finishing_times_hashmap[tasks[i]] = times[i]

    # do the regular bfs traversing
    queue = deque()
    # add the nodes with indegree == 0
    for node in indegree_hashmap:
        if indegree_hashmap[node] == 0:
            queue.append(node)

    while queue:
        popped_node = queue.popleft()

        # we calculate the finish time of the neighbor
        # note: for any neighbor, we can say that the finish time would be
        # the time required to complete its own task + the time required by the previous nodes to finish their tasks
        for neighbor in neighbors_hashmap[popped_node]:
            finishing_times_hashmap[neighbor] = max(finishing_times_hashmap[neighbor], times_hashmap[neighbor]+finishing_times_hashmap[popped_node])

            indegree_hashmap[neighbor] -= 1
            if indegree_hashmap[neighbor] == 0:
                queue.append(neighbor)

    min_completetion_time = max(finishing_times_hashmap.values(), default=0)

    return min_completetion_time
```
