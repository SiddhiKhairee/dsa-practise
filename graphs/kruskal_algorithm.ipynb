### Kruskal's Algorithm

Step 1: Initialize everything
1. Create a list containing all edges and their weights.
2. Sort the edges in ascending order of weight.
3. Initially, treat every node as belonging to its own separate group.
4. Set total_cost = 0.
5. Set edges_used = 0.
Step 2: Iterate through the sorted edges
For each edge, get its two endpoints (u, v) and its weight.
Step 3: Check whether the nodes are already connected
Use Union-Find to determine whether u and v belong to the same group.
- If they belong to the same group, skip the edge because it would create a cycle.
- If they belong to different groups, we can safely add the edge.
Step 4: Add the valid edge
1. Merge the groups containing u and v.
2. Add the edge's weight to total_cost.
3. Increment edges_used by 1.
Step 5: Check whether the MST is complete
A spanning tree with V vertices must contain exactly V − 1 edges.

If edges_used == V - 1, stop.

Otherwise, continue examining the remaining edges.
Step 6: Return the result
If we've selected V − 1 edges, return total_cost.

If we run out of edges before that, the graph is disconnected, so a spanning tree doesn't exist.

```python
graph = {
    "A": [("B", 4), ("C", 2)],
    "B": [("A", 4), ("C", 1), ("D", 5)],
    "C": [("A", 2), ("B", 1), ("D", 8)],
    "D": [("B", 5), ("C", 8)]
}
def kruskal_algo(graph):
    # from this graph, we will have to find a MST
    
    # Step 1: make a parent hashmap:
    parent = {}
    for node in graph:
        parent[node] = node

    # find function
    def find(node):
        if parent[node]!=node:
            parent[node] = find(parent[node])
        return parent[node]

    # union function
    def union(node1, node2):
        root_of_node1 = find(node1)
        root_of_node2 = find(node2)
        if root_of_node1 == root_of_node2: # they are already in the same group
            return False
        else:
            parent[root_of_node2] = root_of_node1 # we combine them and make a group
            return True


    edges = []
    for node in graph:
        for neighbor, weight in graph[node]:
            edges.append((weight, node, neighbor))

    sorted_edges = sorted(edges)
    total_cost = 0
    edges_visited = 0

    for edge in sorted_edges:
        weight, u, v = edge

        # check if u and v are in a group
        created_a_union = union(u,v)
        # if created_a_union is true, it means that they were not in a group and we have visited a new edge
        # if created_a_union is falsem it means they were already in a group, and we do not need to explore the edge
        if created_a_union:
            total_cost += weight
            edges_visited += 1

    if edges_visited == len(graph)-1:
        return total_cost
    else:
        return -1
        
        
            
        
```
