### Task Scheduling

For this problem, given a list of tasks and a list of requirements, compute a sequence of tasks that can be performed, such that we complete every task once while satisfying all the requirements.

Each requirement will be in the form of a list [a, b], where task a needs to be completed first before task b can be completed,

There is guaranteed to be a solution.


Example 1

Input: tasks = ["a", "b", "c", "d"]

requirements = [["a", "b"], ["c", "b"], ["b", "d"]]

Output: ["a", "c", "b", "d"]

```python
from collections import deque
```

```python
def task_scheduling(tasks: list[str], requirements: list[list[str]]) -> list[str]:
    # WRITE YOUR BRILLIANT CODE HERE

    # we start off by making and indegree hashmap function
    def indegrees(tasks, requirements):
        indegree_hsmp = {}

        # start off by making all the keys in the hashmap with the indegree value as 0
        for node in tasks: # every element of the tasks list can be considered as a node
            indegree_hsmp[node] = 0
        
        for element in requirements:
            indegree_hsmp[element[1]] += 1 
        
        return indegree_hsmp

    # we still don't know what the neighbors are for each node, we know their indegrees but we dont know the neighbors explicitly
    # so we make a neighbors hashmap, which will have all the tasks(nodes) as keys, and all the neighbors they are pointing to as values
    neighbors = {}
    for node in tasks:
        neighbors[node] = [] # every node must have a value and that value as of now is an empty list
    # now we traverse through the requirements list
    # for every element in the requirements list, element[0] will be our hashmap key, and element[1] will be appended to the list
    for element in requirements:
        i = neighbors[element[0]]
        i.append(element[1]) # to the list we append the neighbor
    # by the end of this, we have a neighbors hashmap that has node: neighbors 

    # now we perform bfs
    queue = deque()
    results = []
    indegree_hsmp = indegrees(tasks, requirements)

    for node in indegree_hsmp:
        if indegree_hsmp[node] == 0:
            queue.append(node)

    # all the nodes with 0 indegree has been appended to the queue, now we peform bfs on this queue
    while queue:
        popped_node = queue.popleft()
        results.append(popped_node)

        # now for this popped node, we try to get its neighbors
        neighbors_of_popped_node = neighbors[popped_node] # we get a list of all the neighbors of the popped node
        for i in neighbors_of_popped_node:
            indegree_hsmp[i] -= 1 # reduce the indegree value
            if indegree_hsmp[i] == 0:
                # add it to the queue
                queue.append(i)


    # by the end of bfs, hopefully, we have all indegrees as 0, hopefully all nodes have entered the queue, hopefully all the nodes have been popped out of queue
    if len(tasks) == len(results):
        return results
```
