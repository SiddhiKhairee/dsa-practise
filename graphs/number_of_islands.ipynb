### Number of Islands


Given an m x n 2D binary grid grid which represents a map of '1's (land) and '0's (water), return the number of islands.

An island is surrounded by water and is formed by connecting adjacent lands horizontally or vertically. You may assume all four edges of the grid are all surrounded by water.

```python
from typing import List
```

```python
from collections import deque
class Solution:
    def numIslands(self, grid: List[List[str]]) -> int:
        
        # lets have a get neighbors function in place

        def get_neighbors(r, c, grid):
            rows = len(grid)
            cols = len(grid[0])

            directions = [
                (-1,0),
                (0, 1),
                (1, 0),
                (0,-1)
            ]
            neighbors = []
            for dr, dc in directions:
                nr = dr + r
                nc = dc + c

                if 0<=nr<rows and 0<=nc<cols:
                    # check if the value of rows and columns is 1
                    if grid[nr][nc] == '1':
                        # if it is 1 then add it to the list of neighbors
                        neighbors.append((nr,nc))

            return neighbors
        visited = ()
        def bfs(r, c, grid): # by the end of bfs, visited will contain all nodes that are reachable from a source node,
            # provided their value is '1'

            # add the node to the queue
            queue = deque([(r,c)])
            visited.add((r,c))

            while queue:
                r,c = queue.popleft()

                for nr, nc in get_neighbors(r, c):
                    # check if neighbors are in visited
                    if (nr,nc) in visited:
                        continue
                    else:
                        queue.append((nr,nc))
                        visited.add((nr,nc))

            

        number_of_islands = 0
        for i in range(len(grid)):
            for j in range(len(grid[0])):
                if grid[i][j] == '1':
                    # check if this is in visited
                    if (i,j) in visited:
                        continue
                    else: 
                        bfs(i,j, grid)
                        number_of_islands += 1
        return number_of_islands



```

### Time and Space Complexity

Time: O(M × N)

Count how many times each cell gets touched:

The outer double loop looks at each cell once. That's M × N looks.
Each cell goes into the queue at most once, because you mark it visited on push.
Each cell comes out of the queue at most once.
Each pop calls get_neighbors, which checks exactly 4 directions. That's a fixed amount of work, not something that grows with the grid.

So each cell costs a small constant amount of work, total about 5 × M × N. Constants get dropped, which leaves O(M × N). The key point is that BFS is called many times, but the visited set stops any cell from being flooded twice, so the BFS calls together never do more than one full pass over the grid.

Space: O(M × N)

visited can hold every cell if the whole grid is "1". That's M × N entries.
The queue holds only the current frontier, which is smaller, but visited already dominates, so the total is O(M × N).
