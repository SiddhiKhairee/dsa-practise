### Course Schedule - II

here are a total of numCourses courses you have to take, labeled from 0 to numCourses - 1. You are given an array prerequisites where prerequisites[i] = [ai, bi] indicates that you must take course bi first if you want to take course ai.

For example, the pair [0, 1], indicates that to take course 0 you have to first take course 1.
Return the ordering of courses you should take to finish all courses. If there are many valid answers, return any of them. If it is impossible to finish all courses, return an empty array.

 

Example 1:

Input: numCourses = 2, prerequisites = [[1,0]]
Output: [0,1]
Explanation: There are a total of 2 courses to take. To take course 1 you should have finished course 0. So the correct course order is [0,1].
Example 2:

Input: numCourses = 4, prerequisites = [[1,0],[2,0],[3,1],[3,2]]
Output: [0,2,1,3]
Explanation: There are a total of 4 courses to take. To take course 3 you should have finished both courses 1 and 2. Both courses 1 and 2 should be taken after you finished course 0.
So one correct course order is [0,1,2,3]. Another correct ordering is [0,2,1,3].
Example 3:

Input: numCourses = 1, prerequisites = []
Output: [0]
 

Constraints:

1 <= numCourses <= 2000
0 <= prerequisites.length <= numCourses * (numCourses - 1)
prerequisites[i].length == 2
0 <= ai, bi < numCourses
ai != bi
All the pairs [ai, bi] are distinct.


```python
from collections import deque
```

```python
class Solution:
    def findOrder(self, numCourses: int, prerequisites: list[list[int]]) -> list[int]:
        # indegree hashmap
        indegree_hashmap = {}
        for i in range(numCourses):
            indegree_hashmap[i] = 0
        for element in prerequisites:
            indegree_hashmap[element[0]] += 1
        
        # neighbors hashmap
        neighbors = {}
        for i in range(numCourses):
            neighbors[i] = []
        for element in prerequisites:
            neighbors[element[1]].append(element[0])

        # perform a topological sort
        queue = deque()
        order_of_courses = []

        for node in indegree_hashmap:
            if indegree_hashmap[node] == 0:
                queue.append(node)

        while queue:
            popped_node = queue.popleft()
            # the moment we pop a node, we first add it to the order of courses
            order_of_courses.append(popped_node)
            # try to find the nieghbors of the popped node
            neighbors_of_popped_node = neighbors[popped_node]
            for neighbor in neighbors_of_popped_node:
                # we first decrease the indegree by 1
                indegree_hashmap[neighbor] -= 1
                if indegree_hashmap[neighbor] == 0:
                    # append it to the queue
                    queue.append(neighbor)
        
        if len(order_of_courses) == numCourses:
            return order_of_courses
        else:
            return []

```
