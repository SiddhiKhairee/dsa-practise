### Union Find 

Union-Find, also called Disjoint Set Union (DSU), is a data structure that helps us keep track of which nodes belong to the same connected group.

It answers two important questions:
1. Find: Which group does a node belong to?
2. Union: Can we combine two groups into one?
This is especially useful in Kruskal's Algorithm, because we need to determine whether adding an edge will create a cycle.

```python
def union_find(graph):

    # we make a parent hashmap
    parent = {}
    for node in graph:
        parent[node] = node

    # find function returns the root node
    def find(node):
        if parent[node] == node:
            return node
        return find(parent[node])

    # union function for 2 nodes a and b
    def union(a, b):
        root_a = find(a)
        root_b = find(b)

        # if roots are equal, it means they are already unionized
        if root_a == root_b:
            return False
        # if roots are not equal, then we make one root node point and other's root by changing in the parent hashmap
        parent[root_b] = root_a # now they are unionized and they have root node as root_a
        return True

```
