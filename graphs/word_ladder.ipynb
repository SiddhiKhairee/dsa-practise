### Word Ladder
A transformation sequence from word beginWord to word endWord using a dictionary wordList is a sequence of words beginWord -> s1 -> s2 -> ... -> sk such that:

Every adjacent pair of words differs by a single letter.
Every si for 1 <= i <= k is in wordList. Note that beginWord does not need to be in wordList.
sk == endWord
Given two words, beginWord and endWord, and a dictionary wordList, return the number of words in the shortest transformation sequence from beginWord to endWord, or 0 if no such sequence exists.

```python
from collections import deque
from string import ascii_lowercase
def word_ladder(begin: str, end: str, word_list: list[str]) -> int:
    # WRITE YOUR BRILLIANT CODE HERE

    # base condition 
    if begin == end:
        return 0
    if end not in word_list:
        return 0
    # take the current word ladder and put it in a set
    words = set(word_list)

    # define a get neighbors function
    def get_neighbors(word, words):
        # this function will have the word, and will return all the neighbors
        # neighbors are those words whch have only one character different from the original word
        neighbors = []
        for i in range(len(word)):
            # we gotta test every character
            for c in ascii_lowercase:
                if c == word[i]:
                    continue
                possible_neighbor = word[:i]+c+word[i+1:] # we replace the ith word
                if possible_neighbor in words: # it means the neighbor exits in the wordlist
                    neighbors.append(possible_neighbor) 
        return neighbors

    # we apply bfs logic
    queue = deque([begin]) # we add the starting word to the queue
    # we add the same word to visited
    visited = {begin}
    distance = 0 # since we are at the start word the distance is 0
    while queue:
        n = len(queue)
        distance += 1

        for _ in range(n):
            word = queue.popleft()
            for w in get_neighbors(word, words):
                if w == end:
                    return distance
                if w not in visited:
                    queue.append(w)
                    visited.add(w)

    return 0

```

```python
from collections import deque
from string import ascii_lowercase
class Solution:
    def ladderLength(self, beginWord: str, endWord: str, wordList: list[str]) -> int:
        
        # if endWord does not exist in the wordlist
        if endWord not in wordList:
            return 0

        words = set(wordList)
        if beginWord not in words:
            words.add(beginWord)

        # our words is a set of all the possible words in the problem including the beginword

        # define a get_neighbors function
        # this function will take in a word
        # all the neighbors will be returned as a list
        # these neighbors would be such that, they only have one character different from the actual word

        def get_neighbors(word, words):
            neighbors = []
            # we take the word
            # each character of the word will be replaced with all the lowercase ascii letters
            # if we make a word that exists in the words set then we append it as the neighbor
            for i in range(len(word)):
                for c in ascii_lowercase:
                    if c == word[i]:
                        continue
                    else:
                        possible_neighbor = word[:i]+c+word[i+1:] # replace the ith character with all other possible characters
                        if possible_neighbor in words: # if the neighbor actually exists in te set, we append
                            neighbors.append(possible_neighbor)
            return neighbors # for every word we return a list of neighbors

        # we now start with a bfs function
        # we make a queue
        queue = deque([beginWord])
        # we make a visited set and add the starting word to it
        visited = {beginWord}
        distance = 1 # when you are at a word, that counts as one
        # perform bfs
        while queue:
            n = len(queue)
            distance += 1 # we are about to explore the next level so we add 1
            for _ in range(n):
                word = queue.popleft()
                # we get the neighbor of this word
                for neighbor in get_neighbors(word, words):
                    # if our neighbor ends up being the target word, we return distance
                    if neighbor == endWord:
                        return distance
                    # we append to the queue and visited
                    if neighbor not in visited:
                        queue.append(neighbor)
                        visited.add(neighbor)

        # if bfs does not return any value, means word ladder not found. return 0
        return 0


```
