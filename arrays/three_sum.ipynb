### Three Sum (Medium)

Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

 

Example 1:

Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
Explanation: 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.
The distinct triplets are [-1,0,1] and [-1,-1,2].
Notice that the order of the output and the order of the triplets does not matter.
Example 2:

Input: nums = [0,1,1]
Output: []
Explanation: The only possible triplet does not sum up to 0.
Example 3:

Input: nums = [0,0,0]
Output: [[0,0,0]]
Explanation: The only possible triplet sums up to 0.

```python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        op = []
        for i in range(len(nums)):
            if nums[i] == nums [i-1] and i!=0:
                continue
            else: 
                element = 0 - nums[i]
                # first element edgecase
                if i == 0:
                    lp = 1
                    rp = len(nums) - 1
                # Last element
                elif i == len(nums) - 1:
                    lp = 0
                    rp = len(nums) - 2
                else:
                    lp = 0
                    rp = len(nums) - 1
                
                while(lp<rp):
                    if i == lp:
                        lp += 1
                    elif i == rp:
                        rp -= 1
                    else: 
                        sum = nums[lp] + nums[rp]

                        # if our sum is smaller than element, we need a larger sum to add up to zero so we move, lp to right
                        if sum < element:
                            lp += 1
                        
                        # if sum is equal to element means all three elements add upto zero\
                        elif sum == element:
                            triplet = [nums[i], nums[lp], nums[rp]]
                            # triplet.sort()
                            op.append(triplet)
                            lp += 1
                            rp -= 1
                            while (lp< rp and nums[lp]==nums[lp-1]):
                                lp+=1
                            while(lp<rp and nums[rp] == nums[rp+1]):
                                rp-= 1
                        else:
                            rp -= 1

        
        return op        
```
