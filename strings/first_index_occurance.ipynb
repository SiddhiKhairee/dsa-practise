### Find the Index of the First Occurrence in a String (Easy)

Given two strings needle and haystack, return the index of the first occurrence of needle in haystack, or -1 if needle is not part of haystack.

Example 1:

Input: haystack = "sadbutsad", needle = "sad"
Output: 0
Explanation: "sad" occurs at index 0 and 6.
The first occurrence is at index 0, so we return 0.
Example 2:

Input: haystack = "leetcode", needle = "leeto"
Output: -1
Explanation: "leeto" did not occur in "leetcode", so we return -1.


```python
a = [1, 2, 3]
b = [1, 2, 3]
if (a == b):
    print('T')
```

```python
needle = 'leeto'
haystack = 'leetcode'
n_list = [ch for ch in needle]
window_size = len(n_list)
h_list = [ ch for ch in haystack]
h_len = len(h_list)

print(n_list, window_size)
print(h_list, h_len)
```

```python
for i in range(h_len - window_size + 1):
    # make a current window
    current_window = h_list[i:i+window_size]
    if(current_window == n_list):
        print(i)
        break

print("-1")
```

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        h_list = [ch for ch in haystack]
        n_list = [ch for ch in needle]
        window_size = len(n_list)
        for i in range(len(h_list) - window_size + 1):
            # make a current window for every iteration
            current_window = h_list[i:i+window_size]

            # check if current_window and needle are same
            if(n_list == current_window):
                return(i)
            

        return -1

        

```

```python
s = Solution()
result = s.strStr('leetcode', 'leeto')
print(result)
```
