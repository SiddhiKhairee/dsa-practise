### Longest Palindrome Substring (Medium)

Given a string s, return the longest palindromic substring in s.

 

Example 1:

Input: s = "babad"
Output: "bab"
Explanation: "aba" is also a valid answer.
Example 2:

Input: s = "cbbd"
Output: "bb"
 

Constraints:

1 <= s.length <= 1000
s consist of only digits and English letters.

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        longest = ''
        for i in range(len(s)):

            #do the even check
            lp = i
            rp = i+1
            stringy = []
            while(lp>=0 and rp<len(s) and s[lp] == s[rp]):
                stringy.insert(0,s[lp])
                stringy.append(s[rp])
                lp -= 1
                rp += 1
            s_temp = ''.join(stringy)
            longest = max(longest , s_temp, key=len)

            #do the odd check
            lp = i
            rp = i
            stringy = []
            stringy.append(s[i])
            lp-=1
            rp+=1
            while(lp>=0 and rp<len(s) and s[lp]==s[rp]):
                stringy.insert(0,s[lp])
                stringy.append(s[rp])
                lp -= 1
                rp += 1
            s_temp = ''.join(stringy)
            longest = max(longest , s_temp, key=len)

        return longest

        
```
