### Reapeated Substring Pattern (Easy)

Given a string s, check if it can be constructed by taking a substring of it and appending multiple copies of the substring together.

 

Example 1:

Input: s = "abab"
Output: true
Explanation: It is the substring "ab" twice.
Example 2:

Input: s = "aba"
Output: false
Example 3:

Input: s = "abcabcabcabc"
Output: true
Explanation: It is the substring "abc" four times or the substring "abcabc" twice.
 

Constraints:

1 <= s.length <= 104
s consists of lowercase English letters.

```python
class Solution:
    def repeatedSubstringPattern(self, s: str) -> bool:
        # find the divisors of length of the substring
        divisors = []
        for i in range(1, len(s)//2+1):
            if(len(s)%i == 0):
                divisors.append(i)

        # every divisor will act as a substring, of that length
        for i in range(len(divisors)):
            window_size = divisors[i]
            current_window = s[:window_size]
            stringy = current_window*(len(s)//window_size)
            if(stringy == s):
                return True

        return False


        
```
