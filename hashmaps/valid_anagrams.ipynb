### Valid Anagram (Easy)
Given two strings s and t, return true if t is an anagram of s, and false otherwise.

 

Example 1:

Input: s = "anagram", t = "nagaram"

Output: true

Example 2:

Input: s = "rat", t = "car"

Output: false

 

Constraints:

1 <= s.length, t.length <= 5 * 104
s and t consist of lowercase English letters.

```python
from typing import List
```

```python
hsmp1 = {'a':1, 'b':2, 'c':3}
hsmp2 = {'c':3, 'b':2, 'a':2}
if(hsmp1 == hsmp2):
    print("t")
else:
    print("f")
```

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:

        # make hashmaps of s and t with letter:frequency
        hsmp_s ={}
        hsmp_t = {}
        for i in range(len(s)):
            if s[i] in hsmp_s:
                hsmp_s[s[i]] += 1
            else:
                hsmp_s[s[i]] = 1

        
        for i in range(len(t)):
            if t[i] in hsmp_t:
                hsmp_t[t[i]] += 1
            else:
                hsmp_t[t[i]] = 1

        if (hsmp_t == hsmp_s):
            return True
        else:
            return False
```

```python
s = Solution()
result = s.isAnagram('car', 'rac')
print(result)
```
