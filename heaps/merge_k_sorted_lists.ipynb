### Merge K sorted Lists
Given k sorted lists of numbers, merge them into one sorted list.

Input: [[1, 3, 5], [2, 4, 6], [7, 10]]

Output: [1, 2, 3, 4, 5, 6, 7, 10]


```python
from collections import deque
```

```python
from heapq import heapify, heappop, heappush
def merge_k_sorted_lists(lists: list[list[int]]) -> list[int]:
    # WRITE YOUR BRILLIANT CODE HERE
    unified_list = []
    for list in lists:
        for i in list:
            unified_list.append(i)

    heapify(unified_list)
    sorted_list = []
    for i in range(len(unified_list)):
        popped_element = heappop(unified_list)
        sorted_list.append(popped_element)

    return sorted_list
```
