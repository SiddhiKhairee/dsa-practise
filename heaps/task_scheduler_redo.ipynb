### Task Scheduler

You are given an array of CPU tasks, each labeled with a letter from A to Z, and a number n. Each CPU interval can be idle or allow the completion of one task. Tasks can be completed in any order, but there's a constraint: there has to be a gap of at least n intervals between two tasks with the same label.

Return the minimum number of CPU intervals required to complete all tasks.

Example 1:

Input: tasks = ["A","A","A","B","B","B"], n = 2

Output: 8

Explanation: A possible sequence is: A -> B -> idle -> A -> B -> idle -> A -> B.

After completing task A, you must wait two intervals before doing A again. The same applies to task B. In the 3rd interval, neither A nor B can be done, so you idle. By the 4th interval, you can do A again as 2 intervals have passed.

Example 2:

Input: tasks = ["A","C","A","B","D","B"], n = 1

Output: 6

Explanation: A possible sequence is: A -> B -> C -> D -> A -> B.

With a cooling interval of 1, you can repeat a task after just one other task.

Example 3:

Input: tasks = ["A","A","A", "B","B","B"], n = 3

Output: 10

Explanation: A possible sequence is: A -> B -> idle -> idle -> A -> B -> idle -> idle -> A -> B.

There are only two types of tasks, A and B, which need to be separated by 3 intervals. This leads to idling twice between repetitions of these tasks.

```python
from collections import deque
```

```python
from heapq import heappop, heappush
class Solution:
    def leastInterval(self, tasks: list[str], n: int) -> int:

        # step 1: we make a hashmap of all the tasks and their frquencies
        task_frequencies = {}
        for task in tasks:
            if task in task_frequencies:
                task_frequencies[task] += 1
            else:
                task_frequencies[task] = 1

        # step 2: we put all of it into a max_heap, we will be dealing with the max frequency task first
        max_heap = []
        for task, freq in task_frequencies.items():
            heappush(max_heap, ((-1)*freq, task)) # push the freq, task in max_heap

        # we need to create an output array
        # in worst case scenario, the size of the output array will be (n+1)*maximum_frequency
        
        max_frequency = max_heap[0][0]
        max_frequency = -max_frequency
        output_arr = ['idle']*((n+1)*max_frequency)
        # the output array is now initialized with idle

        # now we follow the same mechanism as the rearrange the string
        pointer = 0
        while max_heap:
            popped_freq, popped_element = heappop(max_heap)
            popped_freq = -popped_freq
            for _ in range(popped_freq):
                output_arr[pointer] = popped_element
                pointer += n+1
                if pointer >= len(output_arr):
                    # pointer = 1
                    for i in range(len(output_arr)):
                        if output_arr[i] == 'idle':
                            pointer = i # reset the pointer to the next idle location
                            break



        for i in range(len(output_arr)-1, -1, -1):
            if output_arr[i] != 'idle':
                output_time = i
                break

        return output_time+1



        
```

```python
from heapq import heappop, heappush
from collections import deque

class Solution:
    def leastInterval(self, tasks: list[str], n: int) -> int:

        # step 1: we make a hashmap of all the tasks and their frquencies
        task_frequencies = {}
        for task in tasks:
            if task in task_frequencies:
                task_frequencies[task] += 1
            else:
                task_frequencies[task] = 1

        # step 2: we put all of it into a max_heap, we will be dealing with the max frequency task first
        # we actually do not care about the task, all we need to care about is the frequency
        max_heap = []
        for freq in task_frequencies.values():
            heappush(max_heap, (-1)*freq) # push the freq, in max_heap

        # Step 3: we make a time variable which will calculate how much time it takes to process all the tasks
        time = 0
        # we will also have a queue. this queue will store the (recently_popped_task_frequency, next_processing_time)
        # the next_processing time will be calculated by time + n
        queue = deque()

        while max_heap or queue: # i either maxheap has tasks or queue has tasks we iterate
            #processing time will be incremented by 1 at each iteration
            time += 1
            if max_heap:
                # pop the most frequent task

                popped_frequency = heappop(max_heap)
                # we decrease the frequency by 1
                popped_frequency = popped_frequency + 1 # we do +1 since popped_frequency is negetive and we need to reduce its value
                # for this popped_frequency, we calculate the next_processing_time by doing time+n
                next_processing_time = time+n
                # we check if the popped_frequency > 0, if it is then add it to the queue
                if -popped_frequency > 0: 
                    queue.append((popped_frequency, next_processing_time))

            # when do we remove from the queue and append it to the max_heap again for processing
            # if queue is non zero and the leftmost queue element has the processing time equal to current time
            # this means that  the element can be added to the heap for processing
            if queue and time == queue[0][1]:
                freq , t = queue.popleft()

                heappush(max_heap, freq)

        return time
```
