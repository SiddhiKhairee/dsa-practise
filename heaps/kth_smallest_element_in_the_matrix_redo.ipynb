### Kth smallest element in the matrix

Given an n x n matrix where each row and each column is sorted in ascending order, return the kth smallest value in the matrix.

The rank is based on sorted order of all n^2 entries, so duplicate numbers count multiple times.

Example:
Input:

Copy
matrix = [
  [ 1,  5,  9],
  [10, 11, 13],
  [12, 13, 15]
],
k = 8,
Output: 13
Note:
You may assume k is always valid, 1 ≤ k ≤ n^2, and 1 <= n <= 1000.

```python
from collections import deque
```

```python
from heapq import heapify, heappop
def kth_smallest(matrix: list[list[int]], k: int) -> int:
    # WRITE YOUR BRILLIANT CODE HERE
    listy = []
    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            listy.append(matrix[i][j])

    heapify(listy)
    for i in range(1, k):
        heappop(listy)

    kth_smallest_element = heappop(listy)
    return kth_smallest_element
```

### Altenate approach for optimized solution

Insert the first element of each row into the min-heap.



Pop the smallest element.



Find the next element in the same row and push it into the heap, if one exists.



Repeat until we've popped k elements.



The kth popped element is our answer.


```python
from heapq import heapify, heappop, heappush
def kth_smallest(matrix: list[list[int]], k: int) -> int:
    # WRITE YOUR BRILLIANT CODE HERE
    min_heap = []

    for i in range(len(matrix)):
        min_heap.append((matrix[i][0], i, 0)) # append the first element of every row, and the position of that element

    heapify(min_heap)

    for i in range(k):
        # popp the smallest element
        smallest_element, smallest_element_r, smallest_element_c = heappop(min_heap)

        # once we pop, we add to the heap, the next element of the same row
        if smallest_element_c+1 < len(matrix):
            heappush(min_heap, (matrix[smallest_element_r][smallest_element_c+1], smallest_element_r, smallest_element_c+1))

    return smallest_element

```
