1. Insertion Operation is in two parts

Part 1: Simply Append an Element

Part 2: Now Bubble up, fix the elements position

```python
# this function will take the heap and the index of the newly added element as the parameters
def bubble_up(heap, i):

    # we know that the top of the heap must be the minimum value
    # so for every element at i, we simply compare with its parent
    # we check if i's parent is smaller, if it is small, keep it
    # if it isnt small, then we swap and switch
    while i>0 and heap[(i-1)//2] > heap[i]:
        # we swap if parent element greater than child
        heap[i] , heap[(i-1)//2] = heap[(i-1)//2] , heap[i]

        # after we swap the elements, we need to make sure the the new indices are also asigned properly, so now i becomes the parent
        i = (i-1)//2
```

```python
def insert(heap, val):
    heap.append(val)
    # Call the bubble up function to fix its position
    bubble_up(heap, len(heap) - 1)
    return True
```

2. Deletion operation in heaps is in two parts as well

Part 1: Get the element at the root and return it, take the last element of the heap and replace it with the first element

Part 2: The bubble down approach

```python
def bubble_down(heap, i):

    # currently at the top of heap is the last element (aka largest element)
    n = len(heap)
    smallest_index = i # we make a new variable called smallest_index which essentially will try to store the index of the smallest element
    left_child_index = 2*i + 1
    right_child_index = 2*i + 2 
    while True:

        # compare the ith element since i == smallest_index to the left_child element of the heap.
        # if the left child is smaller, then our smallest_index is updated
        if i < n and heap[left_child_index] < heap[smallest_index]:
            smallest_index = left_child_index
        if i < n and heap[right_child_index] < heap[smallest_index]:
            smallest_index = right_child_index
        # the only way the loop breaks is my ensuring that the smallest index element is actually i
        if i == smallest_index:
            break

        # at every conditional branch we are switching the smallest_index based
        # but everytime we do this we also gotta ensure that the heap values are changed, so that heap[i], over here, heap[0] remains the minimu value
        heap[i] , heap[smallest_index] = heap[smallest_index], heap[i]
        # at every interation we update i
        i = smallest_index
    
```

```python
def delete_min(heap):
    root = heap[0]
    x = heap.pop()
    heap[0] = x
    bubble_down(heap, 0)
    return root
```
