### Median of a data stream

Given a stream of numbers, find the median number at any given time (accurate to 1 decimal place).

Example:


Copy
add_number(1)
add_number(2)
add_number(3)
get_median() == 2.0
add_number(4)
get_median() == 2.5

### Brute Force Approach

```python
from heapq import heappop, heappush
class MedianOfStream:
    def __init__(self):
        # we create a heap
        self.heap_popping = []
        self.heap_continuous = []
    def add_number(self, num: float) -> None:
        # WRITE YOUR BRILLIANT CODE HERE

        # we simply push into the heaps
        heappush(self.heap_continuous, num)
        heappush(self.heap_popping, num)
        return None
   
    def get_median(self) -> float:
        # ALSO HERE

        n = len(self.heap_continuous)
        arr = []
        if self.heap_continuous is not None:
            while self.heap_popping:
                popped_element = heappop(self.heap_popping)
                arr.append(popped_element)

        n = len(arr)
        if n % 2 == 0:
            pos1 = (n-1) // 2
            pos2 = n // 2

            median = (arr[pos1] + arr[pos2]) / 2

        else:
            pos = n // 2
            median = arr[pos]

        self.heap_popping = self.heap_continuous.copy()

        return median
```

### optimized approach

```python
class MedianFinder:

    def __init__(self):
        # we make two heaps
        #  left heap, small heap (will be a max heap)
        # right heap, large heap (will be a min heap)
        self.small_left_max_heap = [] # stores negetive values
        self.large_right_min_heap = [] # stores positive values      

    def addNum(self, num: int) -> None:

        # when we get a number, by default push it into the left heap
        heappush(self.small_left_max_heap, -num)

        # we then check 2 conditons
        # condition 1: Is the largest number in small_left_max_heap greater than smallest number of large_right_min_heap
        # if that is the case, then we pop the largest from left and push it to the right
        if self.small_left_max_heap and self.large_right_min_heap:
            if (-1)*self.small_left_max_heap[0] > self.large_right_min_heap[0]:
                val = (-1)*heappop(self.small_left_max_heap)
                heappush(self.large_right_min_heap, val)

        # our next condition is
        # condition 2: we ensure that the length of both the heaps at most differ by 1

        # left heap larger
        if len(self.small_left_max_heap) > len(self.large_right_min_heap) + 1:
            # remove max from left push to right
            val = (-1)*heappop(self.small_left_max_heap)
            heappush(self.large_right_min_heap, val)

        # right heap larger
        if len(self.large_right_min_heap) > len(self.small_left_max_heap) + 1:
            # remove min from right, push it to the left
            val = heappop(self.large_right_min_heap)
            heappush(self.small_left_max_heap, (-1)*val)

        

    def findMedian(self) -> float:
        if not self.small_left_max_heap and not self.large_right_min_heap:
            raise ValueError("No numbers have been added")

        # uneven length 
        # if left is larger
        if len(self.small_left_max_heap)  > len(self.large_right_min_heap):
            return (-1)*self.small_left_max_heap[0]
        # if right is larger
        if len(self.large_right_min_heap) > len(self.small_left_max_heap):
            return self.large_right_min_heap[0]
        # if lengths are eq
        if len(self.small_left_max_heap) == len(self.large_right_min_heap):
            return ((-1)*self.small_left_max_heap[0] + self.large_right_min_heap[0]) / 2
```
