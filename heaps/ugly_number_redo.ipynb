### Ugly_number

Write a program to find the n-th ugly number.

Ugly numbers are positive numbers whose prime factors only include 2, 3, 5.

Example 1:
Input:n = 10
Output: 12
Explanation: 1, 2, 3, 4, 5, 6, 8, 9, 10, 12 is the sequence of the first 10 ugly numbers.
Note:
1 is typically treated as an ugly number.
n does not exceed 1690.

```python
from collections import deque
```

```python
from heapq import heappop, heappush
def nth_ugly_number(n: int) -> int:
    # WRITE YOUR BRILLIANT CODE HERE

    # instead of caring for prime factorization
    # we start from 1, we keep on multiplying it by 2,3,5
    # add all of them to the heap
    # whatever is the smallest element will always be on the top of the heap, that will be multipled by 2,3,5
    # we keep on dping this for n-1 times
    # at the (n-1)th iteration the nth ugly number is on the top of the list

    # step 1: make a heap, make a visited_number set
    heap = [1]
    visited_numbers = {1}
    # make a multiplication list, which will contain the 2,3,5
    multipliers = (2,3,5)

    for _ in range(n-1):
        popped_val = heappop(heap)
        # multiply this popped_val with the multipliers and add it to heap and the vistited set
        for multiplier in multipliers:
            new_val = popped_val*multiplier
            if new_val not in visited_numbers: 
                heappush(heap, new_val)
                visited_numbers.add(new_val)

    # by the end of the iterations, the topmost value will be the smallest value
    return heap[0] 
```
