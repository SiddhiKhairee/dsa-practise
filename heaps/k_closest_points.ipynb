### K-closest Points
Given a list of points on a 2D plane. Find k closest points to the origin (0, 0).

Input: [(1, 1), (2, 2), (3, 3)], 1

Output: [(1, 1)]


```python
from typing import List
```

```python
from math import sqrt
from heapq import heapify, heappop
def k_closest_points(points: list[list[int]], k: int) -> list[list[int]]:
    # WRITE YOUR BRILLIANT CODE HERE

    # for every point we cal its distance from the origin and maintain that in a distance
    # so we make a distance hashmap
    # where every point_number: distance from the origin
    distance_hsmp = {}
    
    for i in range(len(points)):
        x_i, y_i = points[i][0], points[i][1]
        distance_from_origin = sqrt(x_i**2 + y_i**2)
        distance_hsmp[i] = distance_from_origin

    # we make a distance arr from distance hsmp
    
    distances = []
    for val in distance_hsmp.values():
        distances.append(val)

    # we heapify the distances so it is arranged in the order of distances closest to the origin
    heapify(distances)

    k_smallest_distances = []

    for i in range(k):
        popped_distance = heappop(distances)
        k_smallest_distances.append(popped_distance)

    output = []

    for distance in k_smallest_distances:
        point_number = next((k for k,v in distance_hsmp.items() if v==distance), None)
        element = points[point_number]
        output.append(element)

    return output
```
