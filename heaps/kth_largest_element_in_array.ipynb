### Kth Largest Element in Array

Find the kth largest element in an unsorted array. Note that it is the kth largest element in the sorted order, not necessarily the kth distinct element.

Example 1:
Input: [3,2,1,5,6,4] and k = 2
Output: 5
Example 2:
Input: [3,2,3,1,2,4,5,5,6] and k = 4
Output: 4
Note:
You may assume k is always valid, 1 ≤ k ≤ array's length.



```python
from collections import deque
```

```python
from heapq import heapify, heappop
def find_kth_largest(nums: list[int], k: int) -> int:
    # WRITE YOUR BRILLIANT CODE HERE
    negetive_list = []
    for num in nums:
        negetive_list.append(num*(-1))
    heapify(negetive_list)
    for i in range(1, k):
        heappop(negetive_list)
    kth_smallest = heappop(negetive_list)


    return kth_smallest*(-1)
```

### Alternate Solution (Better Complexity)

Instead of making a heap of n-elements we make a heap only of k-elements

1. push k elements into the heap
2. for k+1 st element in the array, we push it into the heap, and remove the smallest element from the heap


```python
from heapq import heapify, heappop, heappush
def find_kth_largest(nums: list[int], k: int) -> int:
    # WRITE YOUR BRILLIANT CODE HERE

    min_heap = []
    for num in nums:
        heappush(min_heap, num)

        if len(min_heap) > k:
            # pop the smallest element
            heappop(min_heap)

    return min_heap[0]
```
