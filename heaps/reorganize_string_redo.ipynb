### Reorganize String

Given a string s, check if the letters can be rearranged so that two characters that are adjacent to each other are not the same.

If possible, output any possible result. If not possible, return the empty string.

Example 1:
Input:s = "aab"
Output: aba
Example 2:
Input:s = "aaab"
Output: ``
Note:
s will consist of lowercase letters and have length in the range [1, 500].

```python
from heapq import heappush, heappop
from collections import Counter
import sys
def reorganize_string(s: str) -> str:
    # WRITE YOUR BRILLIANT CODE HERE

    # step 1, create a hashmap of characters:frequency
    freq_hsmp = {}
    for i in range(len(s)):
        if s[i] in freq_hsmp:
            freq_hsmp[s[i]] += 1
        else:
            freq_hsmp[s[i]] = 1

    # now we make a heap. In the heap, we store (-frequency, character)
    heap = []
    for character, frequency in freq_hsmp.items():
        heappush(heap, (-frequency, character))

    # now we define a base condition, let n be length of the string
    # number of even elements will be (n+1) // 2
    # number of odd elements is (n//2)
    # consider the most frequent character, if this character appears more times than the number of even positions in the list then no matter what we do we cannot rearrange the string in a way that 2 same characters cannot be together
    # compare topmost frequency with number of even stops
    n = len(s)
    if heap[0][0]*(-1) > (n+1)//2:
        return ''

    # lets say if that is not the case
    # then we traverse down the heap
    # for every element present in the heap, we pop. this ensures that we start from the most frequent element and get it out of the way
    # we pop the element from the heap, and keep on placing the element on even spots, once we are done with that element, we pop the next element from the heap and place it on the odd(remaining) spots
    # keep doing this until the heap is empty
    pointer = 0
    output_string: list[str] = [None]*n
    while heap:
        popped_freq, popped_character = heappop(heap)
        popped_freq = (-1)*popped_freq
        for _ in range(popped_freq):
            output_string[pointer] = popped_character
            pointer += 2 # update pointer by 2, aka we place the characters on every alternate location
            # say pointer ends up becoming greater than the length of string, we then shift the pointer to location 1, and start altenating from there
            if pointer>=n:
                pointer = 1





    return "".join(output_string)
```
