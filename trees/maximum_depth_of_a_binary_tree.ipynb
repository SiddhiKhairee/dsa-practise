### LC 104: Maximum Depth of a Binaray Tree (Easy)
Max depth of a binary tree is the longest root-to-leaf path. Given a binary tree, find its max depth. Here, we define the length of the path to be the number of edges on that path, not the number of nodes.

```python
class Node:
    def __init__(self, val=0, left=None, right= None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def max_depth_of_binary_tree(self, root:Node) -> int:

        # we initialize the number of edges traversed to 0
        number_of_edges = 0

        # we define a dfs function, and this function will traverse down the tree, 
        # everytime it encounters a child node, it will keep a track of th number of edges it has traversed
        def dfs(root, number_of_edges):

            # start off with all the base cases
            # case 1: there is no node 
            if root is None:
                return number_of_edges
            # case 2: we have a node but it is a leaf node
            if root.left is None and root.right is None:
                return number_of_edges
            # case 3: a node only has left subtree, no right child at all
            # in such a case, we add 1 to the number of edges since we have atleast 1 child, and call the recursive dfs function again
            if root.left is not None and root.right is None:
                return dfs(root.left, number_of_edges+1)
            # case 4: same as above, only right subtree exists
            if root.left is None and root.right is not None:
                return dfs(root.right, number_of_edges+1)

            # we have covered up all our base cases
            # our normal case is that node has both the subtrees present
            left_output = dfs(root.left, number_of_edges+1)
            right_output = dfs(root.right, number_of_edges+1)
            # dfs function rn,returns the number of edges
            # we try to ensure that the maximum number of edges is given out as an output
            max_depth_of_binary_tree = max(left_output, right_output)

            return max_depth_of_binary_tree

        # our dfs returns the maximum depth
        # so we call the dfs function on the root
        return dfs(root, number_of_edges)
```

```python
class Solution:
    def maxDepth(self, root: Node | None) -> int:

        # we define a dfs function that only takes a root argument
        def dfs(root):
            # if the node does not exist
            if root is None:
                return 0
            # otherwise we recursively traverse the left and the right subtree
            left = dfs(root.left)
            right = dfs(root.right)
            # now we have to ensure that dfs function returns some value
            # everytime we traverse and we discover a new child node, we add 1 for the current node we are standing on
            return max(left, right) + 1

        # to return maximum edges traversed number-of-nodes - 1
        if root is None:
            return 0 
        else:
            return dfs(root) - 1

        
```
