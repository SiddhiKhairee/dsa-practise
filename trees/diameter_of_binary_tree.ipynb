### Diameter of a binary tree (easy)
Given the root of a binary tree, return the length of the diameter of the tree.

The diameter of a binary tree is the length of the longest path between any two nodes in a tree. This path may or may not pass through the root.

The length of a path between two nodes is represented by the number of edges between them.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def diameterOfBinaryTree(self, root: TreeNode) -> int:

        #we first create a variable best which will keep a track of the best distance between the leftmost subtree and the rightmost subtree
        best = 0
        # define a dfs function
        def dfs(root):
            nonlocal best
            # base case, if the node does not exist, no best diameter
            if root is None:
                return 0

            # we call dfs on the left and right subtree
            left = dfs(root.left)
            right = dfs(root.right)
            # keep a track of the best
            best = max((left+right), best)
            # our dfs function does not return best though
            # it returns the height of the subtrees
            # we simply keep a track of the best 
            return (max(left, right) + 1)

        # we call dfs on the root
        dfs(root)
        return best
```

Why dfs returns height. Every node needs two numbers from its children: how tall the left side is and how tall the right side is. The only way a child can pass that up is through its return value. That's left and right. Without it, the parent can't compute anything.

Why best isn't passed up. The diameter at node 2 (best = 2) is not useful for node 1. Node 1 doesn't care how long the best path inside node 2's subtree was. It only cares how tall node 2's subtree is, because a path through node 1 goes down into that subtree. So best goes into a side variable that every node can update and nobody has to pass around.
