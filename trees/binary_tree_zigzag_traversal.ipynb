### Binary tree Zigzag Traversal


```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
from collections import deque
```

```python
class Solution:
    def zigzagLevelOrder(self, root: TreeNode | None) -> list[list[int]]:
        # we try to first write code of normal level order traversal

        # for zigzag traversing, we maintain a flag, if flag = true, append normally, if flag is false append in reverse
        res = []
        flag = True
        if root is None:
            return res
        # we enqueue the root node
        queue = deque([root])
        # now we traverse down the tree
        # at each level we dequeue the nodes, and as soon as we do that we add their children
        while len(queue) > 0:
            n = len(queue)
            level = []

            for i in range(n):
                # whatever nodes are present in the current queue, we remove them and add them to the level list
                node = queue.popleft()
                # add it to level
                level.append(node.val)
                # now we add the appended nodes children to the queue
                for child in [node.left, node.right]:
                    if child is not None:
                        queue.append(child)
            # by the end of the for loop we have a level of nodes that have been traversed and dequeued from the queue, we add these levels to the result
            # this is where flag comes in picture
            if flag:
                res.append(level)
            else:
                res.append(level[::-1])
            # change the flag value at each iteration
            flag = not flag

        return res
```

### Time and space complexity

Time stays O(n) and space stays O(n).
