### Binary Tree Minimum depth

Given a binary tree, find its minimum depth.

The minimum depth is the number of nodes along the shortest path from the root node down to the nearest leaf node.

Note: A leaf is a node with no children.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
from collections import deque
```

```python
class Solution:
    def minDepth(self, root: TreeNode | None) -> int:

        #we start off by defining depth as -1 for no root
        depth = -1
        if root is None:
            return depth
        # if root exists we add it to the queue
        queue = deque(root)
        while len(queue) > 0:
            n = len(queue)
            depth += 1
            for i in range(n):
                # we dequeue the parent
                node = queue.popleft()
                # we encode the children
                if node.left is None and node.right is None:
                    # we encounter a leaf, so we return depth
                    return depth
                if node.left is not None: # encountered a left subtree
                    queue.append(node.left)
                if node.right is not None: 
                    queue.append(node.right)
        return depth
```
