### Lowest Common Ancestor of BST
Given a binary search tree (BST), find the lowest common ancestor (LCA) node of two given nodes in the BST.

According to the definition of LCA on Wikipedia: “The lowest common ancestor is defined between two nodes p and q as the lowest node in T that has both p and q as descendants (where we allow a node to be a descendant of itself).”

 

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
class Solution:
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':

        # we know that the tree is a bst
        # assume that the node is not none
        if root is not None:
            # then what we do is we compare the value of p and q with the root

            # if both p and q are less than root, recursively check left subtree
            if p.val < root.val and q.val < root.val:
                return self.lowestCommonAncestor(root.left, p, q)
            # if both are greater than the root, recursively check the right subtree
            if p.val > root.val and q.val > root.val:
                return self.lowestCommonAncestor(root.right, p, q)

            # if any one of the values are equal, this implies that the root is the LCA
            if p.val == root.val or q.val == root.val:
                return root
            # this condition is the state where root lies in the range between p and q, thus p and q seperate at the root
            # return root as LCA
            else:
                return root

            
```

### Brute Force
1. Do a DFS to find the path from the root to p, recording every node you pass through.
2. Do the same for q.
3. Walk both paths from the top, together. The last node they share is the LCA. After that point they diverge.


For p = 6, q = 2:

Path to 6: [3, 5, 6]

Path to 2: [3, 5, 2]

They match at 3, they match at 5, and then 6 ≠ 2. The last shared node is 5.

For p = 6, q = 1:

Path to 6: [3, 5, 6]

Path to 1: [3, 1]

They match at 3, then 5 ≠ 1. The last shared node is 3.

### Time and Space Complexity

Time is O(h). 

Each step moves down one level and does O(1) work, so the cost is the height of the tree. On a balanced tree, h = log n, so it's O(log n), which is where your "dividing by half" comes from. On a skewed tree, h = n, so it's O(n), because you're not actually halving anything. A BST isn't guaranteed to be balanced, so the safe statement is "O(h), which is O(log n) if balanced."


Space for your recursive version is O(h), from the call stack. That's right.
