### Binary Search Tree 

left node < Node < right node

```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def find_in_binary_search_tree(self, root, value) -> bool:

        # base case
        # if the root does not exist return false
        if root is None:
            return False
        # if root exists, and the value matches
        if root.val == value:
            return True
        # if root.val is greater than value, search in left subtree
        elif root.val > value:
            self.find_in_binary_search_tree(root.left, value)
        else:
            self.find_in_binary_search_tree(root.right, value)

    def insert_in_binary_search_tree(self, root, value) -> Node|None:

        # base case if node does not exist
        if root is None:
            return Node(value)

        # now we compare, if root.val is greater than value, then we go to left subtree
        if root.val > value:
            left_subtree = self.insert_in_binary_search_tree(root.left, value)
        else:
            right_subtree = self.insert_in_binary_search_tree(root.right, value)

        return root


        
```
