### Valid Binary Search Tree

```python
from math import inf
```

```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def valid_binary_search_tree(root:None) -> bool:
    # we start off with a dfs function to explore the tree\

    # to make sure that something is a binary search tree, the nodes need to be within a particular range
    def dfs(root, min_range, max_range):
        if root is None: # so the range is -inf to inf
            return (True)

        # if root exists but is outside the range
        if root.val <= min_range or root.val >=  max_range:
            return False

        # otherwise we recursively call the dfs on left and right subtree
        # now for left subtree, the max_range value should be the "root" val
        # similarly for right subtree, the min_range must be the "root" val
        return dfs(root.left, min_range, root.val) and dfs(root.right, root.val, max_range)

    # now we call this dfs function on the root, and give the initial root node range as -inf to inf
    return dfs(root, -inf, inf)
```

3. Can the tree be empty, and is an empty tree a valid BST?

### Notes

Brute force: for every node, check that the max of its left subtree is less than the node and the min of its right subtree is greater, and that both subtrees are valid too. It's O(n·h) time, which is O(n²) on a skewed tree and O(n log n) on a balanced one, with O(h) space.

Optimal: pass (lo, hi) bounds down, strict on both sides. It's O(n) time and O(h) space, and O(h) is O(log n) balanced and O(n) skewed.

Follow-up answers:

Duplicates go left: change one operator. node.val >= hi becomes node.val > hi. The lower bound stays strict.
Java Integer.MIN_VALUE trap: a node whose value is exactly Integer.MIN_VALUE or MAX_VALUE gets rejected by the strict check against the sentinel bounds. Fix it with Long bounds, or with null bounds.
1M-deep skewed tree: recursion overflows the call stack. Use an explicit stack of (node, lo, hi) in a loop. Big-O stays O(h), but the memory moves to the heap, so it doesn't crash.
