### Pre order traversal in a binary tree

we first visit a particular node -> then traverse the left subtree -> then visit the right subtree

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right 
```

```python
def pre_order_traversal(root:Node) -> None:
    # base case if no node at root, we stop
    if root is None:
        return None
    # get the value of the root first
    print(root.val)
    # traverse down the left subtree
    pre_order_traversal(root.left)
    #traverse down the right subtree
    pre_order_traversal(root.right)
```
