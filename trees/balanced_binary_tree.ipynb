### Balanced Binary Tree (Easy)
Given a binary tree, determine if it is height-balanced.

 

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isBalanced(self, root: TreeNode | None) -> bool:
        tree = root
        def dfs(tree):
            # we define base case as
            # if node does not exist, we return true, because tree is balanced, but we also return -1 since there is no height to calculate
            if tree is None:
                return (True, -1)

            # in all the other cases we return T/F or the maximum depth of the subtree
            left_is_balanced, left_height = dfs(tree.left)
            right_is_balanced, right_height = dfs(tree.right)
            # if our left and right are balanced and
            # the difference between the left subtree height and the right subtree height is less than 1
            # this implies that our tree is balanced and we return true for that
            if left_is_balanced and right_is_balanced and abs(left_height-right_height)<=1:
                tree_balanced = True
            else:
                tree_balanced = False
            return(tree_balanced, max(left_height, right_height)+1)
        # we call the dfs function and return its output
        is_balanced_tree, _ = dfs(root)
        return is_balanced_tree
```
