### Depth First Search

this is basically pre-order traversal
first visit the node -> we traverse to the left subtree -> traverse to the right subtree

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def dfs(root:Node, target:Node):

    # base case is: our root node does not exist
    if root is None:
        return None
    # another base case, if our root is target
    if root == target:
        return root
    # we traverse down the left subtree
    left = dfs(root.left, target)
    if left is not None: # the reason why we include this line is because, if we find the target then we dont need to explore the right subtree
        return left
    right = dfs(root.right, target)
    return right
```
