### Invert Binary Tree (Easy)
Given the root of a binary tree, invert the tree, and return its root.

```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def invertTree(self, root: Node | None) -> Node | None:

        # we start of by making a dfs function, that would traverse from the root to the leaves of a tree
        def dfs(root):
            # our base case is that, if root is none, we return none
            if root is None:
                return None
            # other than that we recursively call dfs on the left subtree and the right subtree
            # a tree is said to be swapped only when its left and right nodes are swapped
            dfs(root.left)
            dfs(root.right)
            root.left, root.right = root.right, root.left

            # this dfs function returns root
            return root
        return dfs(root)
```
