### Inorder tranversal in trees

if you are on a node,
 1. Visit left branch first -> then visit the node -> then visit the right branch

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def in_order_traversal(root: Node) -> None:

    # we basically give the function the root node
    # if root is not None yet, then we call the same function on the left branch
    if root is not None:
        in_order_traversal(root.left)
        # whatever output we get from the left branch is printed
        print(root.val)
        # then we move onto call the right branch
        in_order_traversal(root.right)
```

```python
def in_order_traversal(root: Node) -> None:

    # base case if root is returning None we end it
    if root is None:
        return None
    # if root has some value in it then 
    # we first visit the left subtree
    in_order_traversal(root.left)
    # then we move onto the actual node, and print its value
    print(root.val)
    # then we move onto the right subtree
    in_order_traversal(root.right)
```
