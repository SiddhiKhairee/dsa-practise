### Binary Tree Right Side View 

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
from collections import deque
```

### Approach 1

```python
class Solution:
    def rightSideView(self, root: TreeNode | None) -> list[int]:
        # we do a normal level order traversal first
        res = []
        if root is None:
            return res
        # if there is a root, we add it to the queue
        queue = deque([root])
        # at each level, we dequeue, as soon as we dequeue a node, we add its children
        while (len(queue) > 0):
            n = len(queue)
            level = []
            for i in range (n):
                # we pop all the nodes on the level and add it to the level array
                node = queue.popleft()
                level.append(node.val)
                for child in [node.left, node.right]:
                    # add child to the queue
                    if child is not None:
                        queue.append(child)
            # we append the level to the result
            res.append(level[-1]) # the only diffference is since we only want the rightmost node we append the last element
        return res
```
