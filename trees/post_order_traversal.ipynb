### Post Order  Traversal
First visit the left subtree -> then visit the right subtree -> then visit the node

```python
class Node:
    def __init__(self, val, left=None, right= None):
        self.val = val
        self.left = None
        self.right = None
```

```python
def post_order_traversal(root:None) -> None:
    # base case, if the node is None then we stop 
    if root is None:
        return None
    # otherwise, we first call the traversal function on the left subtree
    post_order_traversal(root.left)
    # then on right subtree
    post_order_traversal(root.right)
    # finally we access the current node
    print(root.val)
```
