### Lowest Common ancestor on Binary Tree

```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def lca_on_bt(root:Node, p:Node, q:Node):

    # if root node does not exist
    if root is None:
        return None
    # if root exists and the value is equal to p or q
    if root.val == p.val or root.val == q.val:
        return root
    # if thats not the case we check the left and right subtree
    left = lca_on_bt(root.left, p, q)
    right = lca_on_bt(root.right, p, q)
    # if both subtrees output something other than none, it means that the root node is part of both subtrees
    # return root node
    if left and right:
        return root
    # if left subtree outputs something other than none, it implies the root has been found in left
    if left:
        return left
    if right:
        return right
    return None
```
