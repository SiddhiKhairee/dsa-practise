### Level order search in Bianry Tree
Given the root of a binary tree, return the level order traversal of its nodes' values. (i.e., from left to right, level by level).

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
from collections import deque
```

```python
class Solution:
    def levelOrder(self, root: TreeNode | None) -> list[list[int]]:
        # we make a result arr
        res = []
        # if root is None, we return the empty array
        if root is None:
            return res

        # else we make a queue with root in it
        queue = deque([root])
        # keep going while the length of the queue is more than 0
        while len(queue) > 0:
            # let n be the number of nodes at the current level
            n = len(queue)
            new_level = [] # every removed node's value is added to this
            for i in range(n):
                # we first dequeue the current node
                root = queue.popleft()
                new_level.append(root.val)
                # now we gotta add roots children to the node
                for child in [root.left, root.right]:
                    # if child is not none, we enqueue the child
                    if child is not None:
                        queue.append(child)
            # we append the new_level list in our result
            res.append(new_level)

        return res
                        
        
```

### Questions to ask to the interviewers

1. Can the tree be empty? What should I return then?
2. Are node values unique, or can they be negative or duplicated?
3. How large can the tree get? Could it be skewed, so depth approaches n?
4. Should each level go strictly left to right?

### Tighter Explanation
I use a queue for BFS. Each while iteration processes one level: I snapshot the queue size as n, pop exactly n nodes, record their values, and enqueue their non-null children. Those children wait in the queue for the next iteration. I append each level's list to the result.

### Brute Force

The brute force I was fishing for:

- Compute the tree's height h with one recursive pass.
- For each d from 0 to h-1, run a recursive function that walks the whole tree and collects only the nodes at depth d.
- Append each of those lists to the result.

### Time Complexity
Time: O(n)

Nested loops don't automatically multiply. Multiplying only works when each loop runs its full length independently. Here they don't. The for loop runs n_level times inside each while iteration, so across the whole run it executes:

(level 1 size) + (level 2 size) + ... = total number of nodes = n

### Space Complexity

Space: O(n), but the queue is not constant

The queue grows and shrinks. At any moment it holds roughly one level of nodes, so its size is the width of the tree. In a perfect binary tree, the bottom level has about n/2 nodes, so the queue peaks near n/2, which is O(n). Separately, the result list holds every value, which is O(n) as well. So you got the right answer, but "the queue is constant" would be marked wrong in an interview. Say "the queue holds at most one level, so O(width), which is O(n) worst case."
